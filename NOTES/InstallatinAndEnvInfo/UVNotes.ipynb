# 1. What is `uv`?

`uv` is a **modern, high-performance Python package and project manager** developed by **Astral**.

It is written in **Rust**, making it significantly faster than traditional Python tools.

Unlike `pip`, `uv` can:

- Create virtual environments
- Install packages
- Manage dependencies
- Run Python programs
- Synchronize project dependencies
- Manage Python versions (in many workflows)

> **Note:** `uv` does **not** have an official full form. It is simply the name of the tool.

---

# 2. Why was `uv` created?

Traditional Python development requires multiple tools.

Example:

```bash
python -m venv .venv
pip install pandas
pip freeze > requirements.txt
python main.py
```

Here we are using:

- `venv`
- `pip`
- `python`

`uv` combines many of these workflows into one fast tool.

Example:

```bash
uv venv
uv add pandas
uv sync
uv run main.py
```

### Benefits of `uv`

- Extremely fast
- Written in Rust
- Simple commands
- Better dependency management
- Ideal for modern Python development
- Widely adopted in new projects

# 5. What is a Virtual Environment?

A **Virtual Environment** is an **isolated Python environment** created specifically for one project.

Each virtual environment contains:

- Its own Python interpreter
- Its own installed libraries
- Its own package versions

This ensures that one project's dependencies do not interfere with another project's dependencies.

---

# 6. Why do we need a Virtual Environment?

Suppose you are working on two different projects.

## Project A

```text
Python 3.12
Pandas 2.2
NumPy 2.0
```

## Project B

```text
Python 3.10
Pandas 1.5
NumPy 1.24
```

If everything is installed globally, your computer looks like this:

```text
Computer

Python

├── Pandas 2.2
├── NumPy 2.0
```

Now imagine Project B requires Pandas 1.5.

If you upgrade Pandas to 2.2 for Project A, Project B may stop working.

This is called a **dependency conflict**.

---

## Solution

Create separate virtual environments.

```text
Computer

Project A
│
├── Python
├── Pandas 2.2
└── NumPy 2.0

Project B
│
├── Python
├── Pandas 1.5
└── NumPy 1.24
```

Each project is completely independent.

# 8. How `uv` Creates a Virtual Environment

When you execute:

```bash
uv venv
```

`uv` creates a standard Python virtual environment inside your project.

Example:

```text
DataScienceProject/

│── .venv/
│── main.py
```

Notice:

```
.venv
```

The dot (`.`) indicates a hidden folder on Linux/macOS and is the modern naming convention.

# Why Do We Use TOML?

TOML is mainly used to store **configuration settings** for applications and projects.

Instead of writing configuration directly inside Python code, we can keep it in a separate configuration file.

### Example

```

name = "my_project"
version = "1.0.0"
author = "Shubham"
```

Python or another tool can read these settings when required.

* * *

# What is `pyproject.toml`?

In modern Python projects, the most important TOML file is:

```

pyproject.toml
```

`pyproject.toml` is a **standard configuration file for Python projects**.

It can contain information about:

*   Project name
*   Project version
*   Project dependencies
*   Python version requirements
*   Build system
*   Development tools
*   Formatting tools
*   Testing tools
*   Package configuration

* * *

# Importance of `pyproject.toml`

## 1\. Stores Project Information

We can define basic information about the project.

Example:

```

[project]
name = "my-data-science-project"
version = "0.1.0"
description = "Data Science Project"
```

This tells tools:

*   What the project is called
*   Which version it is
*   What the project is about

* * *

# 2\. Manages Dependencies

We can specify the Python packages required by the project.

Example:

```

[project]
dependencies = [
    "numpy",
    "pandas",
    "matplotlib",
    "scikit-learn"
]
```

Now the project's dependencies are clearly defined in one place.

Instead of manually remembering:

```

pip install numpy
pip install pandas
pip install matplotlib
pip install scikit-learn
```

the project configuration tells the package manager what is required.

* * *

# 3\. Specifies Python Version

We can specify which Python versions the project supports.

Example:

```

[project]
requires-python = ">=3.10"
```

This means:

> The project requires Python 3.10 or a compatible newer version.

* * *

# 4\. Works with `uv`

Since you are learning **uv**, this is particularly important.

`uv` can use `pyproject.toml` to manage project dependencies and configuration.

For example:

```

[project]
name = "data-science-project"
version = "0.1.0"

dependencies = [
    "numpy",
    "pandas",
    "scikit-learn"
]
```

Then `uv` can use this project configuration to manage the environment and dependencies.

* * *

# 5\. Keeps Project Configuration in One Place

Without a project configuration file, information may be scattered across multiple files and commands.

For example:

```

Python Version
      ↓
Dependencies
      ↓
Project Name
      ↓
Build Configuration
      ↓
Tool Configuration
```

`pyproject.toml` can bring much of this project configuration together.

* * *

# 6\. Configuration for Development Tools

`pyproject.toml` can also contain configuration for tools such as:

*   Ruff
*   Black
*   pytest
*   mypy
*   Hatch
*   Poetry
*   uv

For example, a tool may have its own section:

```

[tool.ruff]
line-length = 88
```

This tells Ruff how the project should be checked.

* * *

# 7\. Reproducibility

One of the most important benefits is **reproducibility**.

Suppose you create a Data Science project on your computer.

Your project requires:

```

numpy
pandas
scikit-learn
matplotlib
```

If these dependencies are defined in the project configuration, another developer can recreate the required environment more easily.

This helps ensure:

> **Same project → Same dependencies → More consistent environment**

* * *

# TOML vs `pyproject.toml`

These two terms should not be confused.

### TOML

TOML is the **file format**.

```

TOML = Configuration file format
```

### `pyproject.toml`

`pyproject.toml` is a **specific TOML file used for Python project configuration**.

```

pyproject.toml
      ↓
Python project configuration
```

Therefore:

> **TOML is a format, while `pyproject.toml` is a Python project configuration file written using TOML syntax.**

* * *

# Example `pyproject.toml`

A simple Data Science project could have:

```

[project]
name = "data-science-project"
version = "0.1.0"
description = "My Data Science Project"
requires-python = ">=3.10"

dependencies = [
    "numpy",
    "pandas",
    "matplotlib",
    "scikit-learn"
]
```

This file tells the Python tooling:

```

Project Name
     ↓
data-science-project

Version
     ↓
0.1.0

Python Requirement
     ↓
Python >= 3.10

Dependencies
     ↓
numpy
pandas
matplotlib
scikit-learn
```

* * *

# TOML Syntax Basics

## Key-Value Pair

```

name = "data-science-project"
version = "1.0.0"
```

Here:

```

name    → Key
"..."   → Value

version → Key
"..."   → Value
```

* * *

## Section

A section is written using square brackets:

```

[project]
```

Example:

```

[project]
name = "my-project"
version = "1.0.0"
```

* * *

## List

Multiple values can be represented using a list:

```

dependencies = [
    "numpy",
    "pandas",
    "scikit-learn"
]
```

* * *

# `pyproject.toml` vs `requirements.txt`

These files can both contain dependency information, but they serve different purposes.

| `pyproject.toml` | `requirements.txt` |
| --- | --- |
| Project configuration | Dependency list |
| Can contain project metadata | Mainly package requirements |
| Can specify Python requirement | Usually doesn't define project metadata |
| Can configure development tools | Generally doesn't |
| Modern Python project standard | Commonly used for dependency installation |
| Used by tools such as uv | Commonly used with pip |
