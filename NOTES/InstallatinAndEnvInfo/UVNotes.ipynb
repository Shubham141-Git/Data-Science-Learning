# 1. What is `uv`?

`uv` is a **modern, high-performance Python package and project manager** developed by **Astral**.

It is written in **Rust**, making it significantly faster than traditional Python tools.

Unlike `pip`, `uv` can:

- Create virtual environments
- Install packages
- Manage dependencies
- Run Python programs
- Synchronize project dependencies
- Manage Python versions (in many workflows)

> **Note:** `uv` does **not** have an official full form. It is simply the name of the tool.

---

# 2. Why was `uv` created?

Traditional Python development requires multiple tools.

Example:

```bash
python -m venv .venv
pip install pandas
pip freeze > requirements.txt
python main.py
```

Here we are using:

- `venv`
- `pip`
- `python`

`uv` combines many of these workflows into one fast tool.

Example:

```bash
uv venv
uv add pandas
uv sync
uv run main.py
```

### Benefits of `uv`

- Extremely fast
- Written in Rust
- Simple commands
- Better dependency management
- Ideal for modern Python development
- Widely adopted in new projects

# 5. What is a Virtual Environment?

A **Virtual Environment** is an **isolated Python environment** created specifically for one project.

Each virtual environment contains:

- Its own Python interpreter
- Its own installed libraries
- Its own package versions

This ensures that one project's dependencies do not interfere with another project's dependencies.

---

# 6. Why do we need a Virtual Environment?

Suppose you are working on two different projects.

## Project A

```text
Python 3.12
Pandas 2.2
NumPy 2.0
```

## Project B

```text
Python 3.10
Pandas 1.5
NumPy 1.24
```

If everything is installed globally, your computer looks like this:

```text
Computer

Python

├── Pandas 2.2
├── NumPy 2.0
```

Now imagine Project B requires Pandas 1.5.

If you upgrade Pandas to 2.2 for Project A, Project B may stop working.

This is called a **dependency conflict**.

---

## Solution

Create separate virtual environments.

```text
Computer

Project A
│
├── Python
├── Pandas 2.2
└── NumPy 2.0

Project B
│
├── Python
├── Pandas 1.5
└── NumPy 1.24
```

Each project is completely independent.

# 8. How `uv` Creates a Virtual Environment

When you execute:

```bash
uv venv
```

`uv` creates a standard Python virtual environment inside your project.

Example:

```text
DataScienceProject/

│── .venv/
│── main.py
```

Notice:

```
.venv
```

The dot (`.`) indicates a hidden folder on Linux/macOS and is the modern naming convention.

---

# 9. Creating a Virtual Environment using `uv`

Move to your project directory.

```bash
cd DataScienceProject
```

Create the environment.

```bash
uv venv
```

A `.venv` folder is created automatically.

---

# 10. Activating the Virtual Environment

## Windows (Command Prompt)

```bash
.venv\Scripts\activate
```

## Windows (PowerShell)

```powershell
.\.venv\Scripts\Activate.ps1
```

## Linux/macOS

```bash
source .venv/bin/activate
```

After activation:

```text
(.venv)

C:\Projects\DataScienceProject>
```

The `(.venv)` prefix means you are working inside the virtual environment.

# 11. Installing Packages
Before Installing package it should have the .toml file so for this use uv init which will create the project structure then use uv add packageName


Install one package.

```bash
uv add pandas
```

Install multiple packages.

```bash
uv add numpy pandas matplotlib seaborn
```

These packages are installed only in the current project's virtual environment.

---

# 12. Managing Dependencies

Synchronize project dependencies.

```bash
uv sync
```

Run your application.

```bash
uv run main.py
```

Install packages from an existing `requirements.txt`.

```bash
uv pip install -r requirements.txt
```
