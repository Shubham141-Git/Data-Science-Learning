# Surrogate Key

## What is a Surrogate Key?

A **Surrogate Key** is an **artificial/generated key** that is created specifically to uniquely identify each row in a table.

It has **no real-world/business meaning**.

A surrogate key is usually generated by the database using:

- Auto-increment
- Identity columns
- Sequences
- UUIDs

---

# Simple Example

Suppose we have an `Employee` table:

| Employee_ID | Name | Email | Department |
|---:|---|---|---|
| 1 | Shubham | shubham@gmail.com | QA |
| 2 | Rahul | rahul@gmail.com | Data Science |
| 3 | Amit | amit@gmail.com | Development |

Here:

```text
Employee_ID → Surrogate Key
```

The values:

```text
1
2
3
```

are generated only to uniquely identify employees.

The number `1` does not have any business meaning.

---

# Why Do We Need a Surrogate Key?

Suppose we try to use `Email` as the primary key:

| Email | Name |
|---|---|
| shubham@gmail.com | Shubham |
| rahul@gmail.com | Rahul |

This can cause problems because business information can change.

For example:

```text
Old Email:
shubham@gmail.com

New Email:
shubham123@gmail.com
```

If email is used as the primary key, changing the email can affect relationships with other tables.

Instead, we can use:

```text
Employee_ID
```

which normally remains unchanged.

---

# Surrogate Key Example Using SQL

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY GENERATED ALWAYS AS IDENTITY,
    name VARCHAR(100),
    email VARCHAR(100),
    department VARCHAR(100)
);
```

The database can automatically generate:

```text
1
2
3
4
5
...
```

when new employees are inserted.

> The exact syntax for automatically generated keys differs between database systems.

---

# Inserting Data

We don't need to provide the `employee_id` if the database generates it.

```sql
INSERT INTO Employee
(name, email, department)
VALUES
('Shubham', 'shubham@gmail.com', 'QA'),
('Rahul', 'rahul@gmail.com', 'Data Science'),
('Amit', 'amit@gmail.com', 'Development');
```

The database may generate:

| employee_id | name | email | department |
|---:|---|---|---|
| 1 | Shubham | shubham@gmail.com | QA |
| 2 | Rahul | rahul@gmail.com | Data Science |
| 3 | Amit | amit@gmail.com | Development |

# Surrogate Key vs Natural Key

A very important interview concept is the difference between a **Surrogate Key** and a **Natural Key**.

---

## Natural Key

A **Natural Key** is a key that comes from real-world/business data and naturally identifies a record.

### Examples

- Email address
- Aadhaar number
- PAN number
- Employee number assigned by a business
- Product code

For example:

```text
email = shubham@gmail.com
```

If the business considers email unique, it could be used as a natural key.

---

## Surrogate Key

A **Surrogate Key** is artificially created by the database/application.

Example:

```text
employee_id = 101
```

The number `101` does not describe anything about the employee.

It simply identifies the employee.

---

# Difference Between Surrogate Key and Natural Key

| Feature | Surrogate Key | Natural Key |
|---|---|---|
| Meaning | Artificial/generated identifier | Real-world/business identifier |
| Business Meaning | No | Yes |
| Generated by | Database/Application | Usually comes from business data |
| Example | `employee_id = 101` | `email = shubham@gmail.com` |
| Usually changes? | Rarely | Can change |
| Size | Usually small | Can be larger |
| Performance | Often efficient for joins | Can be less efficient if large |
| Complexity | Simple | Can be complex |

---

# Surrogate Key vs Primary Key

These two terms are related but **not the same**.

### Primary Key

A **Primary Key** is a constraint that uniquely identifies each row.

### Surrogate Key

A **Surrogate Key** describes the **nature/origin of the key** — it is an artificially generated identifier.

Therefore:

```text
Surrogate Key
      +
Primary Key
```

can be used together.

### Example

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    name VARCHAR(100)
);
```

Here:

```text
employee_id
     ↓
Surrogate Key
     +
Primary Key
```

The column is a **surrogate key** because it was artificially generated, and it is a **primary key** because it uniquely identifies each row.

---

# Example with Two Tables

Consider:

### Employee Table

| employee_id | name | department_id |
|---:|---|---:|
| 101 | Shubham | 1 |
| 102 | Rahul | 2 |
| 103 | Amit | 1 |

### Department Table

| department_id | department_name |
|---:|---|
| 1 | QA |
| 2 | Data Science |

Here:

```text
Employee
   |
   | employee_id → Surrogate Primary Key
   |
   | department_id → Foreign Key
   ↓
Department
```

The `employee_id` uniquely identifies each employee.

---

# Advantages of Surrogate Keys

## 1. Simple

Usually an integer:

```text
1, 2, 3, 4...
```

is easier to work with than a long business key.

---

## 2. Stable

A surrogate key generally does not change even when business information changes.

Example:

```text
employee_id = 101

Email:
old@gmail.com
      ↓
new@gmail.com
```

The employee still has:

```text
employee_id = 101
```

---

## 3. Efficient for Joins

Integer surrogate keys are generally compact and efficient for indexing and joins.

Example:

```sql
SELECT *
FROM Employee e
JOIN Department d
ON e.department_id = d.department_id;
```

---

## 4. Independent of Business Rules

The key does not depend on business information.

For example:

```text
employee_id = 101
```

does not depend on:

- Employee name
- Email
- Department
- Salary

---

# Disadvantages of Surrogate Keys

## 1. No Business Meaning

Looking at:

```text
employee_id = 101
```

does not tell us anything about the employee.

---

## 2. Additional Column

The database needs an additional column to store the surrogate key.

---

## 3. Does Not Automatically Guarantee Business Uniqueness

For example:

```text
employee_id = 101
email = abc@gmail.com
```

The surrogate key is unique, but we may still need:

```sql
UNIQUE(email)
```

to ensure emails are unique according to the business rule.

---

# Real-World Example

Imagine an e-commerce system.

### Product Table

| product_id | product_code | product_name |
|---:|---|---|
| 1 | LAP001 | Laptop |
| 2 | MOB001 | Mobile |
| 3 | TAB001 | Tablet |

Here:

```text
product_id
    ↓
Surrogate Key
```

while:

```text
product_code
    ↓
Natural/Business Key
```

The `product_id` is used internally to identify the product.

The `product_code` has meaning to the business.

# Surrogate Key in Data Warehousing

Surrogate keys are **very common in Data Warehousing**.

For example:

### Customer Dimension

| customer_key | customer_id | customer_name |
|---:|---|---|
| 1 | CUST001 | Shubham |
| 2 | CUST002 | Rahul |
| 3 | CUST003 | Amit |

Here:

```text
customer_key
     ↓
Surrogate Key
```

and:

```text
customer_id
     ↓
Business/Natural Key
```

This distinction is especially important in **Data Science, Data Engineering, and Data Warehousing interviews**.

# Important Interview Questions

## Q1. What is a surrogate key?

### Answer

> A surrogate key is an artificially generated identifier used to uniquely identify a row in a table. It has no business meaning and is commonly implemented using an auto-incrementing integer, identity column, sequence, or UUID.

---

## Q2. Give an example of a surrogate key.

### Answer

```text
Employee_ID = 101
```

If `Employee_ID` is generated only to uniquely identify the employee and has no business meaning, it is a surrogate key.

---

## Q3. What is the difference between a surrogate key and a natural key?

### Answer

> A surrogate key is artificially generated and has no business meaning, whereas a natural key comes from real-world business data and has meaning.

Example:

```text
Surrogate Key → employee_id = 101

Natural Key → email = shubham@gmail.com
```

---

## Q4. Is a surrogate key always a primary key?

### Answer

> No. A surrogate key is a type of identifier, while a primary key is a database constraint. However, a surrogate key is very commonly used as the primary key of a table.

---

## Q5. Why are surrogate keys used in databases?

### Answer

Surrogate keys are used because they are:

- Simple
- Stable
- Usually compact
- Independent of business rules
- Efficient for relationships and joins

---

## Q6. Can a surrogate key change?

### Answer

> Normally, a surrogate key should remain stable once assigned. It should not change when business attributes such as name, email, or address change.

---

## Q7. Why are surrogate keys important in Data Warehousing?

### Answer

> Surrogate keys provide stable identifiers for dimension records and help manage historical changes in dimension data. They also separate the warehouse's internal identifiers from business keys coming from source systems.

---

# Super Key

A **Super Key** is a set of **one or more columns** that can **uniquely identify a row** in a table.

The important point is:

> **A Super Key can contain extra/unnecessary columns.**

---

# Example

Consider an `Employee` table:

| employee_id | email | name | department |
|-------------|-------|------|------------|
| 101 | shubham@gmail.com | Shubham | QA |
| 102 | rahul@gmail.com | Rahul | Data Science |
| 103 | amit@gmail.com | Amit | Development |

Assume:

- `employee_id` is unique
- `email` is unique

Then the following are **Super Keys**:

```text
{employee_id}
{email}
{employee_id, name}
{employee_id, department}
{email, name}
{email, department}
{employee_id, email}
```

Why?

Because each of these combinations can uniquely identify an employee.

---

# Why is `{employee_id, name}` a Super Key?

Consider:

```text
{employee_id, name}
```

`employee_id` alone is already enough to identify the employee.

So:

```text
employee_id → uniquely identifies row
```

Adding `name` is unnecessary, but the combination still uniquely identifies the row.

Therefore:

```text
{employee_id, name}
        ↓
    Super Key
```

### Key Idea

> **A Super Key may contain extra columns.**

---

# Candidate Key

A **Candidate Key** is a **minimal Super Key**.

This means:

> It uniquely identifies a row, and you cannot remove any column from it without losing uniqueness.

For our Employee table:

```text
{employee_id}
{email}
```

can be Candidate Keys.

Why?

Because both:

- Uniquely identify an employee
- Do not contain unnecessary columns

---

# Super Key vs Candidate Key

Suppose:

```text
{employee_id, name}
```

is a Super Key.

But:

```text
{employee_id}
```

is already enough to uniquely identify the employee.

Therefore:

```text
{employee_id, name}
        ↓
    Super Key
        |
        | Remove unnecessary column
        ↓
{employee_id}
        ↓
 Candidate Key
```

---

# Primary Key

A **Primary Key** is the Candidate Key that is selected as the main key for the table.

For example:

```text
Candidate Keys:

{employee_id}
{email}

        ↓

Choose employee_id

        ↓

Primary Key = employee_id
```

Therefore:

```text
Super Key
    ↓
Candidate Key
    ↓
Primary Key
```

### Important

> **Not every Super Key is a Candidate Key.**

> **Not every Candidate Key is the Primary Key.**

---

# Relationship Between Super Key, Candidate Key and Primary Key

```text
                 SUPER KEYS
                /          \
               /            \
     Candidate Keys       Extra Columns
            |
            ↓
      Primary Key
```

A more precise representation:

```text
Super Key
   |
   ├── Minimal → Candidate Key
   |                 |
   |                 └── Selected → Primary Key
   |
   └── Non-Minimal → Super Key only
```

---

# Example with Student Table

Consider:

| student_id | email | name | city |
|-------------|-------|------|------|
| 1 | a@gmail.com | Amit | Pune |
| 2 | b@gmail.com | Rahul | Mumbai |
| 3 | c@gmail.com | Shubham | Delhi |

Assume:

- `student_id` is unique
- `email` is unique

### Super Keys

```text
{student_id}
{email}

{student_id, name}
{student_id, city}
{email, name}
{email, city}

{student_id, email}
{student_id, email, name}
{student_id, email, name, city}
```

All of these can uniquely identify a student.

### Candidate Keys

```text
{student_id}
{email}
```

These are minimal Super Keys.

### Primary Key

Suppose we choose:

```text
student_id
```

Then:

```text
student_id → Primary Key
```

---

# Super Key vs Candidate Key vs Primary Key

| Feature | Super Key | Candidate Key | Primary Key |
|---------|------------|---------------|-------------|
| Uniquely identifies row | ✅ | ✅ | ✅ |
| Can contain extra columns | ✅ | ❌ | ❌ |
| Must be minimal | ❌ | ✅ | ✅ |
| Number in a table | Many | One or more | One |
| Can be selected as Primary Key | ❌ | ✅ | Already selected |
| Example | `{id, name}` | `{id}` | `id` |

---

# Super Key vs Surrogate Key

These two concepts are **different**.

## Super Key

A Super Key describes a **set of columns that uniquely identifies a row**.

Example:

```text
{employee_id, name}
```

## Surrogate Key

A Surrogate Key describes an **artificially generated identifier**.

Example:

```text
employee_id = 101
```

A surrogate key can also be a Candidate Key and Primary Key.

For example:

```text
employee_id
     ↓
Surrogate Key
     ↓
Candidate Key
     ↓
Primary Key
```

# Important Interview Questions

## Q1. What is a Super Key?

### Answer

> A Super Key is a set of one or more attributes that can uniquely identify a row in a table. It may contain additional attributes that are not necessary for uniqueness.

---

## Q2. Give an Example of a Super Key

Consider:

```text
Employee(employee_id, name, email)
```

If `employee_id` is unique:

```text
{employee_id}
```

is a Super Key.

Also:

```text
{employee_id, name}
```

is a Super Key because `employee_id` already uniquely identifies the row.

---

## Q3. What is the Difference Between Super Key and Candidate Key?

### Answer

> A Super Key can contain unnecessary attributes, whereas a Candidate Key is a minimal Super Key.

Example:

```text
{employee_id, name} → Super Key

{employee_id}       → Candidate Key
```

---

## Q4. Is Every Candidate Key a Super Key?

### Answer

> **Yes.** Every Candidate Key is a Super Key because it uniquely identifies a row. However, a Super Key is not necessarily a Candidate Key because it may contain unnecessary attributes.

```text
Candidate Key ⊂ Super Key
```

Conceptually, Candidate Keys are the **minimal Super Keys**.

---

## Q5. Can a Table Have Multiple Super Keys?

### Answer

> Yes. A table can have many Super Keys because different combinations of attributes can uniquely identify a row.

---

## Q6. Can a Table Have Multiple Candidate Keys?

### Answer

> Yes. A table can have multiple Candidate Keys. One of them is selected as the Primary Key.

Example:

```text
Candidate Keys:

employee_id
email
```

We could choose:

```text
employee_id → Primary Key
```

---

## Q7. Is a Primary Key a Super Key?

### Answer

> **Yes.** A Primary Key is a minimal Super Key selected to uniquely identify rows in a table.

---

## Q8. Can a Super Key Contain Multiple Columns?

### Answer

> Yes. A Super Key can contain one or more columns.

Example:

```text
{employee_id}
```

or:

```text
{department_id, employee_number}
```

The second is a **composite Super Key** if the combination uniquely identifies a row.

# Alternate Key

An **Alternate Key** is a **Candidate Key that is not selected as the Primary Key**.

This is an important database interview concept.

---

# Simple Example

Suppose we have an `Employee` table:

| employee_id | email | name | department |
|-------------|-------|------|------------|
| 101 | shubham@gmail.com | Shubham | QA |
| 102 | rahul@gmail.com | Rahul | Data Science |
| 103 | amit@gmail.com | Amit | Development |

Assume:

- `employee_id` is unique
- `email` is unique

Therefore, we have two **Candidate Keys**:

```text
employee_id
email
```

Suppose we select:

```text
employee_id
```

as the **Primary Key**.

Then:

```text
employee_id → Primary Key
email       → Alternate Key
```

Why?

Because `email` was a Candidate Key but was **not selected as the Primary Key**.

---

# Relationship Between Keys

```text
Super Keys
     ↓
Candidate Keys
     ↓
   ┌───────────────┐
   ↓               ↓
Primary Key    Alternate Key
   ↓               ↓
Selected        Remaining
Candidate       Candidate
Key             Keys
```

### Easy Way to Remember

```text
Candidate Keys
      ↓
Choose one
      ↓
Primary Key

Remaining Candidate Keys
      ↓
Alternate Keys
```

---

# Example

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    email VARCHAR(100) UNIQUE,
    name VARCHAR(100)
);
```

Here:

```text
employee_id
     ↓
Primary Key
```

And:

```text
email
  ↓
UNIQUE
  ↓
Candidate Key
  ↓
Not selected as Primary Key
  ↓
Alternate Key
```

---

# Alternate Key vs Primary Key

| Primary Key | Alternate Key |
|---|---|
| Selected Candidate Key | Candidate Key not selected |
| Main identifier of the table | Other possible unique identifier |
| Only one primary key constraint per table | There can be multiple alternate keys |
| Cannot be NULL | NULL handling depends on how the alternate key is implemented |
| Example: `employee_id` | Example: `email` |

> **Important:** An alternate key is a conceptual database-design term. In SQL, it is commonly implemented using a `UNIQUE` constraint.

---

# Example with Student Table

Consider:

| student_id | email | phone | name |
|---|---|---|---|
| 1 | a@gmail.com | 9876543210 | Amit |
| 2 | b@gmail.com | 9876543211 | Rahul |
| 3 | c@gmail.com | 9876543212 | Shubham |

Assume:

```text
student_id → Unique
email      → Unique
phone      → Unique
```

Therefore:

### Candidate Keys

```text
student_id
email
phone
```

Suppose we select:

```text
student_id
```

as the Primary Key.

Then:

```text
student_id → Primary Key

email      → Alternate Key
phone      → Alternate Key
```

So a table can have **multiple Alternate Keys**.

---

# Alternate Key vs UNIQUE Key

These terms are related but not exactly the same.

## Alternate Key

Describes the **role of a Candidate Key** that wasn't chosen as the Primary Key.

## UNIQUE Constraint

A SQL constraint that enforces uniqueness.

Example:

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    email VARCHAR(100) UNIQUE
);
```

Here:

```text
employee_id → Primary Key

email → Candidate Key / Alternate Key
        +
        UNIQUE constraint
```

---

# Alternate Key vs Super Key vs Candidate Key

| Key | Meaning |
|---|---|
| **Super Key** | Can uniquely identify a row; may contain extra columns |
| **Candidate Key** | Minimal Super Key |
| **Primary Key** | Candidate Key selected as the main key |
| **Alternate Key** | Candidate Key not selected as the Primary Key |

### Example

```text
Employee Table

employee_id
email
name
```

Assume `employee_id` and `email` are unique.

```text
{employee_id, name}
        ↓
    Super Key

{employee_id}
{email}
        ↓
   Candidate Keys

{employee_id}
        ↓
    Primary Key

{email}
        ↓
   Alternate Key
```

## Q1. What is an Alternate Key?

### Answer

> An Alternate Key is a Candidate Key that is not selected as the Primary Key. It can still uniquely identify a row in the table.

---

## Q2. Give an Example of an Alternate Key.

Suppose:

```text
Candidate Keys:

employee_id
email
```

If:

```text
employee_id → Primary Key
```

then:

```text
email → Alternate Key
```

---

## Q3. Can a Table Have Multiple Alternate Keys?

### Answer

> **Yes.** A table can have multiple Candidate Keys, and all Candidate Keys that are not selected as the Primary Key are Alternate Keys.

Example:

```text
Candidate Keys:

employee_id
email
phone
```

If `employee_id` is the Primary Key:

```text
email → Alternate Key
phone → Alternate Key
```

---

## Q4. Is an Alternate Key the Same as a Candidate Key?

### Answer

> Not exactly. An Alternate Key **is a Candidate Key**, but specifically one that was **not selected as the Primary Key**.

```text
Every Alternate Key → Candidate Key

Every Candidate Key → NOT necessarily Alternate Key
```

---

## Q5. Is an Alternate Key the Same as a UNIQUE Constraint?

### Answer

> No. An Alternate Key is a **key concept/design role**, while `UNIQUE` is a **SQL constraint** used to enforce uniqueness.

Example:

```sql
email VARCHAR(100) UNIQUE
```

The `UNIQUE` constraint ensures duplicate email values are not allowed.

---

## Q6. Can an Alternate Key Contain Multiple Columns?

### Answer

**Yes.**

An Alternate Key can be a **composite Candidate Key**.

For example:

```text
Candidate Keys:

student_id
(country_code, phone_number)
```

If `student_id` is selected as the Primary Key:

```text
(country_code, phone_number)
            ↓
      Alternate Key
```

---

## Q7. What Happens If There Is Only One Candidate Key?

If a table has only one Candidate Key and it is selected as the Primary Key:

```text
Candidate Key
      ↓
Primary Key
```

Then there are **no Alternate Keys**.

---

### Answer

> **2 Alternate Keys**

# Foreign Key

A **Foreign Key** is a column or set of columns in one table that **references a Primary Key or a suitable Unique Key in another table**.

Its main purpose is to:

- Create relationships between tables
- Maintain **Referential Integrity**
- Prevent invalid relationships between tables

---

# Simple Example

Suppose we have two tables.

## 1. Department Table

| department_id | department_name |
|---:|---|
| 1 | QA |
| 2 | Data Science |
| 3 | Development |

Here:

```text
department_id → Primary Key
```

---

## 2. Employee Table

| employee_id | name | department_id |
|---:|---|---:|
| 101 | Shubham | 1 |
| 102 | Rahul | 2 |
| 103 | Amit | 1 |

Here:

```text
department_id → Foreign Key
```

The `department_id` in the `Employee` table refers to:

```text
Department.department_id
```

### Relationship

```text
Department
    |
    | department_id (Primary Key)
    |
    ↓
Employee
    |
    | department_id (Foreign Key)
```

---

# SQL Example

## Parent Table

```sql
CREATE TABLE Department
(
    department_id INT PRIMARY KEY,
    department_name VARCHAR(100)
);
```

## Child Table

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    name VARCHAR(100),
    department_id INT,

    FOREIGN KEY (department_id)
        REFERENCES Department(department_id)
);
```

Here:

```text
Department
    ↓
Parent Table

Employee
    ↓
Child Table
```

---

# Parent Table and Child Table

When a Foreign Key is involved, we commonly refer to two tables as:

## Parent Table

The table containing the referenced key.

```text
Department
```

## Child Table

The table containing the Foreign Key.

```text
Employee
```

Therefore:

```text
Department (Parent)
       |
       | Primary Key
       ↓
Employee (Child)
       |
       | Foreign Key
```

---

# Why Do We Use Foreign Keys?

The main purpose is to maintain **Referential Integrity**.

## What is Referential Integrity?

Referential Integrity means:

> A Foreign Key value in the child table should refer to a valid key value in the parent table, unless the Foreign Key is allowed to be `NULL`.

For example, suppose the Department table contains:

| department_id | department_name |
|---:|---|
| 1 | QA |
| 2 | Data Science |

Then this is valid:

```text
Employee.department_id = 1
```

because department `1` exists.

But this is invalid:

```text
Employee.department_id = 10
```

if department `10` does not exist.

---

# Example of Foreign Key Constraint

Suppose we try:

```sql
INSERT INTO Employee
(employee_id, name, department_id)
VALUES
(104, 'Raj', 10);
```

If department `10` does not exist in the `Department` table, the database will generally reject the insert because of the Foreign Key constraint.

# Foreign Key Does Not Have to Have the Same Name

The column names don't have to be identical.

### Department Table

```sql
CREATE TABLE Department
(
    id INT PRIMARY KEY,
    department_name VARCHAR(100)
);
```

### Employee Table

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    name VARCHAR(100),
    dept_id INT,

    FOREIGN KEY (dept_id)
        REFERENCES Department(id)
);
```

Here:

```text
Employee.dept_id
       ↓
Department.id
```

The names are different, but the relationship is valid.

# Foreign Key vs Primary Key

| Primary Key | Foreign Key |
|---|---|
| Uniquely identifies a row | Creates a relationship between tables |
| One primary key constraint per table | A table can have multiple Foreign Keys |
| Cannot contain NULL | Can contain NULL unless `NOT NULL` is specified |
| Values must be unique | Values can be duplicated |
| Usually located in the parent table | Usually located in the child table |
| Example: `Department.department_id` | Example: `Employee.department_id` |

---

# Important Point: Foreign Key Values Can Repeat

This is very important for interviews.

Consider:

| employee_id | name | department_id |
|---:|---|---:|
| 101 | Shubham | 1 |
| 102 | Rahul | 2 |
| 103 | Amit | 1 |
| 104 | Priya | 1 |

Notice:

```text
department_id = 1
```

appears multiple times.

This is perfectly valid because a Foreign Key **does not have to be unique**.

It means:

```text
Many Employees
      ↓
Department 1
```

This represents a **one-to-many relationship**.

---

# Foreign Key Can Be NULL

A Foreign Key can generally contain `NULL` unless the column has a `NOT NULL` constraint.

Example:

| employee_id | name | department_id |
|---:|---|---:|
| 101 | Shubham | 1 |
| 102 | Rahul | NULL |

Here, Rahul does not currently have a department assigned.

This is allowed if:

```sql
department_id INT
```

is nullable.

If we write:

```sql
department_id INT NOT NULL
```

then `NULL` is not allowed.

# Foreign Key and Relationships

Foreign Keys are commonly used to create relationships between tables.

## One-to-Many Relationship

Example:

```text
Department
    |
    | 1
    |
    |───────────< Employee
                 Many
```

One department can have many employees.

Example:

```text
QA Department
     |
     ├── Shubham
     ├── Amit
     ├── Priya
     └── Rahul
```

---

# Foreign Key with JOIN

Foreign Keys are frequently used when joining tables.

```sql
SELECT
    e.name,
    d.department_name
FROM Employee e
JOIN Department d
    ON e.department_id = d.department_id;
```

Result:

| name | department_name |
|---|---|
| Shubham | QA |
| Rahul | Data Science |
| Amit | QA |

Here:

```text
e.department_id
       =
d.department_id
```

allows us to connect the two tables.

---

# ON DELETE

A Foreign Key can define what should happen when a referenced parent row is deleted.

Common options include:

- `CASCADE`
- `SET NULL`
- `RESTRICT`
- `NO ACTION`

### Example

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    name VARCHAR(100),
    department_id INT,

    FOREIGN KEY (department_id)
        REFERENCES Department(department_id)
        ON DELETE CASCADE
);
```

### ON DELETE CASCADE

If a department is deleted:

```text
Department 1
    ↓
Employees belonging to Department 1
    ↓
Automatically deleted
```

> Use cascading deletes carefully because they can remove many child records.

---

# ON UPDATE

Some databases allow actions such as:

```sql
ON UPDATE CASCADE
```

This means that if the referenced key changes, the related Foreign Key values can be updated automatically.

Example:

```sql
FOREIGN KEY (department_id)
REFERENCES Department(department_id)
ON UPDATE CASCADE;
```

---

# Composite Foreign Key

A Foreign Key can consist of **multiple columns**.

Example:

```sql
FOREIGN KEY (student_id, course_id)
REFERENCES Enrollment(student_id, course_id);
```

Here:

```text
student_id + course_id
        ↓
Composite Foreign Key
```

This is used when the relationship depends on a combination of columns.

# Important Interview Questions ⭐

## Q1. What is a Foreign Key?

### Answer

> A Foreign Key is a column or set of columns in one table that references a key in another table. It is used to establish relationships between tables and maintain referential integrity.

---

## Q2. What is the difference between Primary Key and Foreign Key?

### Answer

> A Primary Key uniquely identifies records within its own table, while a Foreign Key is used to reference a key in another table and establish a relationship between the tables.

---

## Q3. Can a Foreign Key contain duplicate values?

### Answer

> **Yes.** A Foreign Key does not have to be unique.

Example:

```text
Employee

employee_id | department_id
------------|-------------
101         | 1
102         | 1
103         | 2
104         | 1
```

Multiple employees can belong to the same department.

---

## Q4. Can a Foreign Key contain NULL?

### Answer

> **Yes**, unless the Foreign Key column has a `NOT NULL` constraint.

Example:

```sql
department_id INT
```

allows `NULL`.

But:

```sql
department_id INT NOT NULL
```

does not allow `NULL`.

---

## Q5. Can a table have multiple Foreign Keys?

### Answer

> **Yes.** A table can have multiple Foreign Keys referencing different tables.

Example:

```text
Order
  |
  ├── customer_id → Customer
  |
  ├── product_id  → Product
  |
  └── employee_id → Employee
```

---

## Q6. Does a Foreign Key have to reference a Primary Key?

### Answer

> Not necessarily. A Foreign Key commonly references a Primary Key, but depending on the database system, it can also reference a column or combination of columns that has a suitable `UNIQUE` constraint.

For interview purposes, remember:

```text
Foreign Key
     ↓
Usually references
     ↓
Primary Key / Unique Key
```

---

## Q7. What is Referential Integrity?

### Answer

> Referential Integrity ensures that a Foreign Key value corresponds to a valid referenced key in the parent table, unless the Foreign Key is allowed to be `NULL`.

---

## Q8. What happens if we try to insert an invalid Foreign Key?

Example:

```text
Department table:

1 → QA
2 → Data Science
```

Trying to insert:

```text
Employee.department_id = 10
```

when department `10` doesn't exist will generally result in a **Foreign Key constraint violation**.

---

## Q9. Can a Foreign Key reference a Foreign Key?

A Foreign Key normally references a candidate/unique key in another table, rather than being defined merely because another column happens to be a Foreign Key.

The important rule is:

```text
Child Foreign Key
       ↓
References a valid key in Parent
```

---

## Q10. What is a Self-Referencing Foreign Key?

A table can have a Foreign Key that references another row in the **same table**.

Example:

```sql
CREATE TABLE Employee
(
    employee_id INT PRIMARY KEY,
    name VARCHAR(100),
    manager_id INT,

    FOREIGN KEY (manager_id)
        REFERENCES Employee(employee_id)
);
```

Here:

```text
Employee
   |
   ├── employee_id
   |
   └── manager_id
            ↓
       Employee.employee_id
```

Example:

| employee_id | name | manager_id |
|---:|---|---:|
| 101 | Shubham | NULL |
| 102 | Rahul | 101 |
| 103 | Amit | 101 |

Here:

```text
Shubham
   |
   ├── Rahul
   └── Amit
```

This is called a **self-referencing Foreign Key**.

---

# Interview Trick Question ⭐

### Q: Is a Foreign Key always unique?

**Answer: No.**

This is a common interview trap.

```text
department_id

1
1
1
2
2
3
```

This is completely valid.

Why?

Because many employees can belong to the same department.

```text
Foreign Key
     ↓
Column in Child Table
     ↓
References Key in Parent Table
     ↓
Creates Relationship
     ↓
Maintains Referential Integrity
```

### Example

```text
Department
-------------------------
department_id (PK)
department_name


Employee
-------------------------
employee_id (PK)
name
department_id (FK)
```

Relationship:

```text
Department
     |
     | department_id
     ↓
Employee
```

### Remember

> **Primary Key → Uniquely identifies a row**

> **Foreign Key → Connects tables**

> **Foreign Key → Can contain duplicate values**

> **Foreign Key → Can contain NULL unless NOT NULL is specified**

> **Parent Table → Contains the referenced key**

> **Child Table → Contains the Foreign Key**

> **Referential Integrity → Keeps relationships between tables valid**
