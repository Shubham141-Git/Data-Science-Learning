# Difference Between Database and DBMS

The easiest way to remember:

> **Database = Where the data is stored**  
> **DBMS = Software used to manage that data**

---

# 1. Database

A **Database** is an organized collection of data.

### Example

Suppose a company stores employee information:

| Emp_ID | Name | Department | Salary |
|--------|------|------------|--------|
| 101 | Shubham | QA | 50,000 |
| 102 | Rahul | Data Science | 70,000 |

This collection of employee data is a **Database**.

### Examples of Databases

- Employee Database
- Banking Database
- Hospital Database
- E-commerce Database

---

# 2. DBMS

**DBMS = Database Management System**

A **DBMS** is software that allows us to:

- Create data
- Store data
- Retrieve data
- Update data
- Delete data
- Manage data

### Examples of DBMS

- MySQL
- PostgreSQL
- Oracle
- Microsoft SQL Server
- SQLite

---

# How Database and DBMS Work Together

```text
              DBMS
                ↓
        ┌───────────────┐
        │   Database    │
        ├───────────────┤
        │ Employees     │
        │ Departments   │
        │ Projects      │
        └───────────────┘
```

The **DBMS acts as an interface between the user/application and the database**.

---

# Database vs DBMS

| Feature | Database | DBMS |
|---------|----------|------|
| Meaning | Collection of data | Software that manages data |
| Purpose | Stores data | Manages the stored data |
| Type | Data | Software |
| Performs Operations? | No, by itself | Yes |
| Security | Contains data | Provides data security |
| Examples | Employee Database, Banking Database | MySQL, PostgreSQL, Oracle |

---

# Simple Real-Life Analogy

Imagine a **Library**.

```text
Library Books → Database
Librarian     → DBMS
```

The **books** contain the information.

The **librarian** helps you:

- Find a book
- Add a book
- Remove a book
- Update information
- Control who can access books

Similarly:

```text
Database → Stores the data

DBMS → Manages the data
```

---

# Interview Answer

If the interviewer asks:

**"What is the difference between a Database and DBMS?"**

You can answer:

> A **database** is an organized collection of data, whereas a **DBMS** is software that allows us to create, store, retrieve, update, delete, and manage that data. For example, an employee database stores employee information, while **MySQL or PostgreSQL** can be used as a DBMS to manage that database.

# Race Codition

A race condition happens when two or more threads try to access or change the same data at the same time. Since we don't know which thread will finish first, the final result can be incorrect or unpredictable.

Or, in an even simpler sentence:

A race condition occurs when multiple threads race to update the same resource at the same time, leading to unexpected results.

Easy Analogy

Imagine two people trying to write on the same whiteboard at the same time. Their writing overlaps, and the final message becomes messy. That's exactly what happens in a race condition when multiple threads modify the same data simultaneously.

One-Line Interview Answer

A race condition is a situation where multiple threads access and modify the same shared data at the same time, causing unpredictable or incorrect results.

# Understanding the ER (Entity Relationship) Diagram

An **ER (Entity Relationship) Diagram** is a visual representation of a relational database. It shows:

- **Entities (Tables)**
- **Attributes (Columns)**
- **Relationships** between tables
- **Primary Keys (PK)** and **Foreign Keys (FK)**

Think of it as a **blueprint of the database**.

---

# Overall Structure

```text
           DEPARTMENT
                |
             (1 : N)
                |
            EMPLOYEE
          /     |      \
      (N:M)   (1:N)   (N:1)
       /         |        \
PROJECT   EMPLOYEE_PROJECT  ADDRESS
                |
           DESIGNATION
```

The **EMPLOYEE** table is the central table, and most other tables are connected to it.

---

# 1. Department Table

```text
+----------------------+
|     DEPARTMENT       |
+----------------------+
| Dept_ID (PK)         |
| Dept_Name            |
| Location             |
+----------------------+
```

## Purpose

The **Department** table stores information about departments in a company.

### Example

| Dept_ID | Dept_Name | Location |
|----------|-----------|----------|
| 1 | QA | Pune |
| 2 | Data Science | Mumbai |

---

## Relationship with Employee

```text
Department
      |
      | 1
      |
      | N
Employee
```

### Meaning

This is a **One-to-Many (1:N)** relationship.

- One department can have many employees.
- One employee belongs to only one department.

### Example

```text
QA Department
    |
    |---- Shubham
    |---- Rahul
    |---- Amit
```

---

# 2. Employee Table

```text
+------------------------+
|      EMPLOYEE          |
+------------------------+
| Emp_ID (PK)            |
| Emp_Name               |
| Email                  |
| Hire_Date              |
| Salary                 |
| Dept_ID (FK)           |
+------------------------+
```

## Purpose

The **Employee** table stores details about employees.

It contains:

- Employee ID
- Name
- Email
- Hire Date
- Salary
- Department ID

---

## Primary Key (PK)

```text
Emp_ID
```

### Meaning

A **Primary Key** uniquely identifies every record in a table.

Example

| Emp_ID | Name |
|--------|------|
|101|Shubham|
|102|Rahul|

There cannot be two employees with the same Employee ID.

---

## Foreign Key (FK)

```text
Dept_ID
```

### Meaning

A **Foreign Key** connects one table with another.

Example

### Department Table

| Dept_ID | Department |
|----------|------------|
|1|QA|
|2|Data Science|

### Employee Table

| Emp_ID | Dept_ID |
|---------|----------|
|101|1|

Employee **101** belongs to the **QA** department.

---

# 3. Address Table

```text
+------------------------+
| ADDRESS                |
+------------------------+
| Address_ID (PK)        |
| Emp_ID (FK)            |
| Address_Line1          |
| Address_Line2          |
| City                   |
| State                  |
| Zip_Code               |
+------------------------+
```

## Relationship

```text
Employee

1 -------- N Address
```

### Meaning

One employee can have multiple addresses.

### Example

```text
Shubham

↓

Permanent Address

↓

Temporary Address
```

---

# 4. Project Table

```text
+--------------------+
| PROJECT            |
+--------------------+
| Project_ID (PK)    |
| Project_Name       |
| Start_Date         |
| End_Date           |
+--------------------+
```

## Purpose

Stores project details.

### Example

| Project_ID | Project Name |
|------------|--------------|
|1|Netflix App|
|2|Banking App|

---

# 5. Employee_Project Table (Junction Table)

```text
+---------------------------+
| EMPLOYEE_PROJECT          |
+---------------------------+
| Emp_ID (FK)               |
| Project_ID (FK)           |
| Role                      |
| Hours                     |
+---------------------------+
```

## Why Do We Need This Table?

Because:

- One employee can work on multiple projects.
- One project can have multiple employees.

This creates a **Many-to-Many (N:M)** relationship.

### Example

```text
Employee A

↓

Project A
Project B
Project C
```

Also,

```text
Project A

↓

Employee A
Employee B
Employee C
```

Since SQL databases cannot directly represent a many-to-many relationship, we create a **junction table**.

### Example

| Emp_ID | Project_ID |
|---------|------------|
|101|1|
|101|2|
|102|2|

This table indicates:

- Employee 101 works on Project 1
- Employee 101 works on Project 2
- Employee 102 works on Project 2

---

# 6. Designation Table

```text
+----------------------+
| DESIGNATION          |
+----------------------+
| Desig_ID (PK)        |
| Desig_Name           |
| Level                |
+----------------------+
```

## Purpose

Stores employee job titles.

### Example

| Desig_ID | Designation |
|-----------|-------------|
|1|QA Engineer|
|2|Senior QA|
|3|Data Scientist|

---

## Relationship

```text
Designation

1

↓

Employee

N
```

### Meaning

One designation can be assigned to many employees.

### Example

```text
Data Scientist

↓

Rahul
Priya
Karan
```

## One (1)

```text
1
```

```text
N
```

Means **multiple records**.

Example

Many Employees

# One-to-Many (1:N)

```text
Department

↓

Employee
Employee
Employee
```

### Meaning

One department can have many employees.

---

# Many-to-One (N:1)

```text
Employee
Employee
Employee

↓

Department
```

### Meaning

Many employees belong to one department.

---

# Many-to-Many (N:M)

```text
Student

↓

↓

Course
```

### Meaning

- One student can enroll in many courses.
- One course can have many students.

---

# What is a Junction Table?

A **Junction Table** is used to resolve a **Many-to-Many** relationship.

Example

```text
Employee

↓

Project A
Project B
Project C
```

and

```text
Project A

↓

Employee A
Employee B
Employee C
```

This relationship cannot be represented directly in a relational database.

Instead, we create a junction table.

### Employee_Project Table

| Emp_ID | Project_ID |
|---------|------------|
|101|1|
|101|2|
|102|1|
|103|3|

This converts a **Many-to-Many** relationship into two **One-to-Many** relationships.

|----------|----------|--------------|---------|
| Department | Employee | 1 : N | One department has many employees |
| Employee | Address | 1 : N | One employee can have many addresses |
| Employee | Project | N : M | Employees work on multiple projects, and projects have multiple employees |
| Employee | Employee_Project | 1 : N | One employee can have many project assignments |
| Project | Employee_Project | 1 : N | One project can have many employee assignments |
| Designation | Employee | 1 : N | One designation can be assigned to many employees |

# Key Interview Points

When explaining an ER Diagram in an interview, always follow this sequence:

1. Explain what an ER Diagram is.
2. Identify all the entities (tables).
3. Mention the Primary Key (PK) of each table.
4. Explain the Foreign Keys (FK) and how they connect tables.
5. Describe the relationship type (1:1, 1:N, N:M).
6. Explain why a junction table is required for many-to-many relationships.
7. Give a real-world example for each relationship.

---

# Quick Revision

- **Entity** → Table
- **Attribute** → Column
- **Record** → Row
- **Primary Key (PK)** → Uniquely identifies each record
- **Foreign Key (FK)** → Connects two tables
- **1:1** → One-to-One
- **1:N** → One-to-Many
- **N:1** → Many-to-One
- **N:M** → Many-to-Many (requires a junction table)
- **Junction Table** → Resolves many-to-many relationships by breaking them into two one-to-many relationships.

# Non-Relational Database (NoSQL)

## What is a Non-Relational Database?

A **Non-Relational Database**, commonly known as a **NoSQL database**, is a database that does not primarily store data in traditional relational tables. No SQL Stores data in Flexible formats instead of rigid tables

NoSQL databases are designed to handle:

- Large volumes of data
- Structured data
- Semi-structured data
- Unstructured data
- Frequently changing data
- Distributed applications

---

# Why Do We Need NoSQL?

Traditional relational databases work very well when data has a fixed structure.

However, modern applications generate different types of data such as:

- JSON
- Images
- Videos
- Social media data
- Logs
- Sensor data
- User activity

This data may not have a fixed structure.

NoSQL databases provide:

- Flexible schema
- High scalability
- High availability
- Fast read/write operations
- Distributed data storage

---

# SQL vs NoSQL

| Feature | Relational Database | Non-Relational Database |
|---------|---------------------|-------------------------|
| Other Name | SQL / RDBMS | NoSQL |
| Storage | Tables | Documents, Key-Value, Columns, Graphs |
| Structure | Fixed Schema | Flexible Schema |
| Data | Mostly Structured | Structured, Semi-structured, Unstructured |
| Relationships | Strongly supported | Depends on database type |
| Scaling | Usually Vertical | Commonly Horizontal |
| Query Language | SQL | Depends on Database |
| Examples | MySQL, PostgreSQL | MongoDB, Redis, Cassandra, Neo4j |

---

# Types of NoSQL Databases

There are four major types of NoSQL databases:

1. **Document Database**
2. **Key-Value Database**
3. **Column-Family Database**
4. **Graph Database**

---

# 1. Document Database

A **Document Database** stores data as documents, usually in a JSON-like format.

### Example

```json
{
    "id": 101,
    "name": "Shubham",
    "age": 25,
    "skills": [
        "Python",
        "SQL",
        "Pandas"
    ]
}
```

Different documents can have different fields.

### Example

```json
{
    "id": 101,
    "name": "Shubham",
    "skills": ["Python", "SQL"]
}
```

Another document can have additional information:

```json
{
    "id": 102,
    "name": "Rahul",
    "age": 28,
    "city": "Pune",
    "skills": ["Python", "Machine Learning"]
}
```

### Examples

- MongoDB
- CouchDB

### Used For

- E-commerce
- Content Management
- Mobile Applications
- APIs
- User Profiles

---

# 2. Key-Value Database

A **Key-Value Database** stores data in the form:

```text
KEY → VALUE
```

### Example

```text
101 → Shubham
102 → Rahul
103 → Amit
```

The key uniquely identifies the value.

### Real-World Example

A website can store a user's session:

```text
session_123 → user_101
```

### Examples

- Redis
- Amazon DynamoDB

### Used For

- Caching
- Session Management
- Real-Time Applications
- Shopping Carts
- Gaming

---

# 3. Column-Family Database

A **Column-Family Database** stores data in column families rather than traditional relational tables.

It is designed to handle very large amounts of distributed data.

### Examples

- Apache Cassandra
- Apache HBase

### Used For

- Big Data
- IoT
- Large-Scale Analytics
- Distributed Applications

---

# 4. Graph Database

A **Graph Database** stores data as:

- **Nodes** → Entities
- **Edges** → Relationships
- **Properties** → Information about nodes or relationships

### Example

```text
Shubham
   |
   | FRIEND
   ↓
Rahul
   |
   | WORKS_AT
   ↓
Company
```

Here:

```text
Shubham → Node
Rahul   → Node
Company → Node

FRIEND   → Relationship
WORKS_AT → Relationship
```

### Examples

- Neo4j
- Amazon Neptune

### Used For

- Social Networks
- Recommendation Systems
- Fraud Detection
- Network Analysis

---

# Important Characteristics of NoSQL

## 1. Flexible Schema

NoSQL databases generally allow documents or records to have different structures.

For example:

```text
User 1

Name
Age
City
```

Another user could have:

```text
User 2

Name
Age
Skills
Projects
```

---

# 2. Horizontal Scaling

NoSQL databases are commonly designed to scale by adding more machines.

```text
Server 1
    +
Server 2
    +
Server 3
    +
Server 4
```

This is called **Horizontal Scaling** or **Scaling Out**.

---

# 3. High Availability

NoSQL databases are often designed to keep applications running even if one server fails.

---

# 4. Distributed Architecture

Data can be distributed across multiple servers.

```text
             Database
                 |
       ┌─────────┼─────────┐
       ↓         ↓         ↓
    Server 1  Server 2  Server 3
```

---

# NoSQL Does NOT Mean "No SQL"

### Interview Question

**What does NoSQL mean?**

### Answer

NoSQL generally means **"Not Only SQL"**.

It does not necessarily mean that SQL is completely absent.

Some NoSQL databases may support SQL-like query languages, but they do not follow the traditional relational database model.

---

# SQL vs NoSQL — Important Interview Concept

### SQL Database

Data is generally stored in tables.

```text
Employee Table

| ID | Name | Age |
|----|------|-----|
| 1  | A    | 25  |
| 2  | B    | 30  |
```

### NoSQL Database

Data may be stored as a document.

```json
{
    "id": 1,
    "name": "Shubham",
    "age": 25
}
```

---

# When Should We Use NoSQL?

Use NoSQL when:

- Data structure changes frequently.
- Very large amounts of data need to be handled.
- High scalability is required.
- High availability is required.
- Data is semi-structured or unstructured.
- Fast read/write operations are required.
- Data is distributed across multiple servers.

---

# When Should We Use SQL?

Use a relational database when:

- Data has a well-defined structure.
- Relationships between tables are important.
- Complex joins are required.
- Strong consistency is required.
- Transactions are critical.

### Example

Banking systems require strong consistency and transactions, so relational databases are commonly used.

---

# Advantages of NoSQL

- Flexible schema
- High scalability
- High availability
- Handles large datasets
- Good for distributed systems
- Suitable for semi-structured and unstructured data
- Fast read/write operations for appropriate workloads

---

# Disadvantages of NoSQL

- Complex relationships can be difficult to manage.
- Different NoSQL databases use different query languages.
- Data consistency models can vary.
- Not always suitable for complex transactions.
- Less standardized than SQL databases.

# NoSQL Interview Questions

## Basic Interview Questions

### Q1. What is a NoSQL database?

**Answer:**

> A NoSQL database is a non-relational database designed to store and manage structured, semi-structured, or unstructured data. It provides flexible schemas and is commonly used for large-scale and distributed applications.

---

### Q2. What does NoSQL stand for?

**Answer:**

> NoSQL generally stands for **Not Only SQL**. It refers to databases that do not primarily follow the traditional relational database model.

---

### Q3. What are the types of NoSQL databases?

**Answer:**

There are four major types:

1. Document Database
2. Key-Value Database
3. Column-Family Database
4. Graph Database

Examples:

```text
Document       → MongoDB
Key-Value      → Redis
Column-Family  → Cassandra
Graph          → Neo4j
```

---

### Q4. What is the difference between SQL and NoSQL?

**Answer:**

> SQL databases store data in structured tables with predefined schemas and are suitable for complex relationships and transactions. NoSQL databases use flexible data models such as documents, key-value pairs, columns, or graphs and are commonly used for scalable distributed applications.

---

### Q5. What is a Document Database?

**Answer:**

> A Document Database stores data as documents, usually in JSON or JSON-like format. Each document can have a different structure. MongoDB is a popular example.

---

### Q6. What is a Key-Value Database?

**Answer:**

> A Key-Value Database stores data as key-value pairs. The key uniquely identifies the associated value. Redis is a popular example.

---

### Q7. What is a Graph Database?

**Answer:**

> A Graph Database stores data as nodes and relationships. It is useful when relationships between entities are very important, such as social networks and recommendation systems.

---

### Q8. What is a Column-Family Database?

**Answer:**

> A Column-Family Database stores data in column families and is designed for large-scale distributed data. Cassandra is a popular example.

# Intermediate Interview Questions

## Q9. Why would you choose NoSQL over SQL?

**Answer:**

> I would choose NoSQL when the application requires flexible schemas, horizontal scalability, high availability, or needs to handle large amounts of semi-structured or unstructured data.

---

## Q10. What is a flexible schema?

**Answer:**

> A flexible schema means that records do not necessarily have to follow exactly the same structure.

For example:

```json
{
    "name": "Shubham",
    "age": 25
}
```

Another document can be:

```json
{
    "name": "Rahul",
    "skills": ["Python", "SQL"],
    "city": "Pune"
}
```

Both documents can exist in the same document database.

---

## Q11. What is horizontal scaling?

**Answer:**

> Horizontal scaling means increasing system capacity by adding more servers or machines instead of increasing the capacity of a single server.

```text
Server 1
Server 2
Server 3
Server 4
```

This is also called **Scaling Out**.

---

## Q12. What is vertical scaling?

**Answer:**

> Vertical scaling means increasing the resources of an existing server, such as CPU, RAM, or storage.

```text
Before:

Server
4 GB RAM

        ↓

After:

Server
32 GB RAM
```

### Difference

```text
Vertical Scaling → Add resources to one server

Horizontal Scaling → Add more servers
```

---

## Q13. Is NoSQL always better than SQL?

**Answer:**

> No. SQL and NoSQL solve different problems. SQL is generally better when strong relationships, complex queries, and transactions are important. NoSQL is useful when flexible schemas, horizontal scalability, and distributed data are important.

---

## Q14. Does NoSQL support relationships?

**Answer:**

> Yes, but relationships are handled differently depending on the NoSQL database. Graph databases explicitly model relationships, while document databases may use embedding or references.

---

## Q15. What is MongoDB?

**Answer:**

> MongoDB is a document-oriented NoSQL database that stores data in flexible JSON-like documents rather than traditional rows and columns.

---

## Q16. Why is MongoDB called a NoSQL database?

**Answer:**

> MongoDB is called a NoSQL database because it does not use the traditional relational table-based model. It stores data as documents in collections.

---

## Q17. What is a collection in MongoDB?

**Answer:**

A **Collection** is similar to a table in a relational database.

```text
SQL              MongoDB

Database    →    Database
Table       →    Collection
Row         →    Document
Column      →    Field
```

---

## Q18. What is a document in MongoDB?

**Answer:**

A **Document** is a record stored in MongoDB in a JSON-like BSON format.

Example:

```json
{
    "name": "Shubham",
    "age": 25,
    "city": "Pune"
}
```

---

## Q19. What is BSON?

**Answer:**

> BSON stands for **Binary JSON**. MongoDB uses BSON to store documents internally. It extends JSON by supporting additional data types.

---

## Q20. What is the difference between a MongoDB collection and a SQL table?

**Answer:**

| SQL | MongoDB |
|-----|---------|
|Table|Collection|
|Row|Document|
|Column|Field|
|Primary Key|`_id`|
|SQL Query|MongoDB Query|

> **SQL = Language used to communicate with a database**

---

# 1. What is a Database?

A **Database** is an organized collection of data.

### Example

An employee database may contain:

| Emp_ID | Name | Department | Salary |
|--------|------|------------|--------|
| 101 | Shubham | QA | 50000 |
| 102 | Rahul | Data Science | 70000 |

The database **stores this information**.

---

# 2. What is SQL?

**SQL = Structured Query Language**

SQL is a **language used to interact with relational databases**.

Using SQL, we can:

- Retrieve data
- Insert data
- Update data
- Delete data
- Create tables
- Modify tables
- Filter data
- Join tables

### Example

```sql
SELECT *
FROM Employee;
```

This SQL query asks the database:

> "Give me all the records from the Employee table."

---

# Database vs SQL

| Feature | Database | SQL |
|---------|----------|-----|
| What is it? | Collection of data | Query language |
| Purpose | Stores data | Communicates with database |
| Type | Data storage | Programming/Query language |
| Used for | Storing and managing data | Retrieving and manipulating data |
| Example | Employee Database | `SELECT * FROM Employee;` |
| Examples | MySQL Database, PostgreSQL Database | SELECT, INSERT, UPDATE, DELETE |

```sql
SELECT Name, Salary
FROM Employee
WHERE Salary > 60000;
```

Here:

- **Database** → Contains the Employee data
- **DBMS** → Manages the database
- **SQL** → Tells the DBMS what data you want

# Core Structure of Relational Database

A **Relational Database** stores data in the form of **Tables**.

The basic structure is:

```text
Database
   |
   ├── Table
   │     |
   │     ├── Columns
   │     |
   │     └── Rows
   │
   └── Table
```

---

# 1. Table

A **Table** is a structured collection of data organized into:

- **Rows** --> represent single record int a table
- **Columns**

### Example

| id | name | department | salary |
|----|------|------------|--------|
| 1 | Shubham | QA | 50000 |
| 2 | Rahul | Data Science | 70000 |

Here:

- `id` → Column
- `name` → Column
- `department` → Column
- `salary` → Column
- Each horizontal entry → Row / Record

---

# 2. How to Create a Table

We use the `CREATE TABLE` command to create a table.

### Syntax

```sql
CREATE TABLE table_name
(
    column_name data_type constraints,
    column_name data_type constraints,
    column_name data_type constraints
);
```

### Example

```sql
CREATE TABLE Employee
(
    id INT,
    name VARCHAR(100),
    department VARCHAR(100),
    salary INT
);
```

When we execute this command, the table named `Employee` is created inside the selected database.

---

# 3. Column

A **Column** represents a particular attribute or property of the data.

For example:

```text
id
name
department
salary
```

Each column has three important components:

```text
Column Definition
       |
       ├── Column Name
       |
       ├── Data Type
       |
       └── Constraints
```

### General Syntax

```sql
column_name data_type constraints
```

Example:

```sql
id INT PRIMARY KEY
```

Here:

- `id` → Column Name
- `INT` → Data Type
- `PRIMARY KEY` → Constraint

---

# 4. Column Name

The **Column Name** identifies what type of information the column stores.

Example:

```sql
id
name
age
salary
department
email
```

### Example

```sql
CREATE TABLE Employee
(
    id INT,
    name VARCHAR(100),
    salary INT
);
```

Here:

| Column Name | Meaning |
|-------------|---------|
| `id` | Employee ID |
| `name` | Employee Name |
| `salary` | Employee Salary |

---

# 5. Data Type

A **Data Type** defines what kind of data can be stored in a column.

### Common SQL Data Types

| Data Type | Used For | Example |
|-----------|----------|---------|
| `INT` | Integer numbers | `25` |
| `DECIMAL` | Decimal numbers | `45000.50` |
| `VARCHAR` | Variable-length text | `"Shubham"` |
| `CHAR` | Fixed-length text | `"IN"` |
| `TEXT` | Large text | `"Long description..."` |
| `DATE` | Date | `'2026-08-09'` |
| `TIME` | Time | `'10:30:00'` |
| `DATETIME` / `TIMESTAMP` | Date + Time | `'2026-08-09 10:30:00'` |
| `BOOLEAN` | True/False | `TRUE` |

> **Note:** Exact data types and their behavior can vary slightly between database systems such as MySQL, PostgreSQL, SQL Server, and Oracle.

---

# 6. Constraints

**Constraints** are rules applied to columns to control what data can be stored.

They help maintain **data accuracy, consistency, and integrity**.

### Common Constraints

- `PRIMARY KEY`
- `FOREIGN KEY`
- `NOT NULL`
- `UNIQUE`
- `CHECK`
- `DEFAULT`

---

# 7. PRIMARY KEY

A **PRIMARY KEY** uniquely identifies each row in a table.

### Example

```sql
CREATE TABLE Employee
(
    id INT PRIMARY KEY,
    name VARCHAR(100),
    salary INT
);
```

Here:

```text
id → PRIMARY KEY
```

### Properties

- Must be unique
- Cannot contain `NULL`
- A table has one primary key constraint
- A primary key can consist of one or multiple columns

### Example

| id | name |
|----|------|
| 101 | Shubham |
| 102 | Rahul |

We cannot have:

| id | name |
|----|------|
| 101 | Amit |

because `101` already exists.

---

# 8. NOT NULL

`NOT NULL` means a column **must have a value**.

### Example

```sql
CREATE TABLE Employee
(
    id INT,
    name VARCHAR(100) NOT NULL,
    salary INT
);
```

Here:

```text
name → Cannot be NULL
```

This is allowed:

```text
Shubham
```

This is not allowed:

```text
NULL
```

---

# 9. UNIQUE

`UNIQUE` ensures that values in a column are not duplicated.

### Example

```sql
CREATE TABLE Employee
(
    id INT PRIMARY KEY,
    email VARCHAR(100) UNIQUE
);
```

The following is not allowed:

| id | email |
|----|-------|
| 1 | a@gmail.com |
| 2 | a@gmail.com |

because the email value must be unique.

---

# 10. DEFAULT

`DEFAULT` provides a value automatically when no value is supplied.

### Example

```sql
CREATE TABLE Employee
(
    id INT PRIMARY KEY,
    name VARCHAR(100),
    country VARCHAR(50) DEFAULT 'India'
);
```

If we insert:

```sql
INSERT INTO Employee (id, name)
VALUES (101, 'Shubham');
```

The database automatically uses:

```text
country = India
```

---

# 11. CHECK

`CHECK` ensures that a value satisfies a specified condition.

### Example

```sql
CREATE TABLE Employee
(
    id INT PRIMARY KEY,
    name VARCHAR(100),
    age INT CHECK (age >= 18)
);
```

This is allowed:

```text
age = 25
```

This is not allowed:

```text
age = 15
```

because:

```text
age >= 18
```

must be true.

---

# 12. FOREIGN KEY

A **FOREIGN KEY** is used to establish a relationship between two tables.

### Department Table

```sql
CREATE TABLE Department
(
    dept_id INT PRIMARY KEY,
    dept_name VARCHAR(100)
);
```

### Employee Table

```sql
CREATE TABLE Employee
(
    id INT PRIMARY KEY,
    name VARCHAR(100),
    dept_id INT,
    FOREIGN KEY (dept_id)
        REFERENCES Department(dept_id)
);
```

Here:

```text
Department
    |
    | dept_id (PK)
    |
    ↓
Employee
    |
    | dept_id (FK)
```

`dept_id` in the `Employee` table refers to `dept_id` in the `Department` table.

---

# 13. Complete Example

We can combine all these concepts.

```sql
CREATE TABLE Employee
(
    id INT PRIMARY KEY,
    name VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE,
    age INT CHECK (age >= 18),
    salary DECIMAL(10,2),
    country VARCHAR(50) DEFAULT 'India'
);
```

Here:

| Column | Data Type | Constraint |
|--------|-----------|------------|
| `id` | `INT` | `PRIMARY KEY` |
| `name` | `VARCHAR(100)` | `NOT NULL` |
| `email` | `VARCHAR(100)` | `UNIQUE` |
| `age` | `INT` | `CHECK` |
| `salary` | `DECIMAL(10,2)` | None |
| `country` | `VARCHAR(50)` | `DEFAULT` |

---

A **Row** represents one complete record in a table.

Example:

| id | name | department | salary |
|----|------|------------|--------|

The complete horizontal entry is one **Row / Record**.

# Row in a Relational Database

## What is a Row?

A **Row** represents a **single record** or **instance** in a table.

Each row contains values for the columns defined in the table.

### Example

Consider an `Employee` table:

| id | name | department | salary |
|----|------|------------|--------|
| 101 | Shubham | QA | 50000 |
| 102 | Rahul | Data Science | 70000 |
| 103 | Amit | Development | 65000 |

Here:

```text
Row 1 → 101, Shubham, QA, 50000
Row 2 → 102, Rahul, Data Science, 70000
Row 3 → 103, Amit, Development, 65000
```

Each horizontal entry is a **Row / Record**.

---

# Row = Record

In a relational database, the terms **Row** and **Record** are generally used interchangeably.

For example:

```text
101 | Shubham | QA | 50000
```

This complete entry represents one employee.

Therefore:

```text
Row = Record
```

---

# What Does a Row Contain?

A row contains a value for each column of the table.

For example:

```text
Employee Table

| id | name | department | salary |
|----|------|------------|--------|
| 101 | Shubham | QA | 50000 |
```

The row contains:

```text
id         → 101
name       → Shubham
department → QA
salary     → 50000
```

So:

> **A row is a collection of related values representing one entity or record.**

---

# Row and Column Relationship

A table is made up of **Rows and Columns**.

```text
                  Columns
          ↓          ↓          ↓
       id         name       salary
       ↓           ↓           ↓
    ┌──────┬────────────┬──────────┐
    │ 101  │  Shubham   │  50000   │ ← Row / Record
    ├──────┼────────────┼──────────┤
    │ 102  │ Rahul      │  70000   │ ← Row / Record
    ├──────┼────────────┼──────────┤
    │ 103  │ Amit       │  65000   │ ← Row / Record
    └──────┴────────────┴──────────┘
```

### Remember

```text
Column → Describes an attribute

Row → Represents one record
```

---

```text
1 | Shubham | 25 | Pune
```

```text
2 | Rahul | 24 | Mumbai
```

# Row and Primary Key

Usually, every row in a relational table should be uniquely identifiable.

A **Primary Key** is used to uniquely identify a row.

Example:

| id (PK) | name | salary |
|---------|------|--------|
| 101 | Shubham | 50000 |
| 102 | Rahul | 70000 |

Here:

```text
101 → Uniquely identifies the first row
102 → Uniquely identifies the second row
```

Therefore:

> **A Primary Key uniquely identifies a record (row) in a table.**

---

# Can Two Rows Have the Same Values?

Yes, depending on the table design.

For example:

| id | name | department |
|----|------|------------|
| 101 | Shubham | QA |
| 102 | Shubham | QA |

The `name` and `department` values are the same, but the rows are different because their `id` values are different.

```text
Row 1 → id = 101
Row 2 → id = 102
```

The Primary Key makes them uniquely identifiable.

---

# NULL Value in a Row

A row can contain `NULL` if the corresponding column allows NULL values.

Example:

| id | name | department | salary |
|----|------|------------|--------|
| 101 | Shubham | QA | NULL |

Here:

```text
salary = NULL
```

means the salary value is **unknown or not provided**.

> `NULL` does not mean `0`.

---

# Inserting a Row

We use the `INSERT` statement to add a new row to a table.

### Example

```sql
INSERT INTO Employee
(id, name, department, salary)
VALUES
(101, 'Shubham', 'QA', 50000);
```

This creates a new row:

| id | name | department | salary |
|----|------|------------|--------|
| 101 | Shubham | QA | 50000 |

---

# Updating a Row

We use the `UPDATE` statement to modify values in an existing row.

### Example

```sql
UPDATE Employee
SET salary = 60000
WHERE id = 101;
```

Before:

| id | name | salary |
|----|------|--------|
| 101 | Shubham | 50000 |

After:

| id | name | salary |
|----|------|--------|
| 101 | Shubham | 60000 |

---

# Deleting a Row

We use the `DELETE` statement to remove rows from a table.

### Example

```sql
DELETE FROM Employee
WHERE id = 101;
```

The row where:

```text
id = 101
```

will be removed.

---

# Selecting Rows

We use `SELECT` to retrieve rows.

### Example

```sql
SELECT *
FROM Employee;
```

This retrieves all rows.

### Selecting Specific Rows

```sql
SELECT *
FROM Employee
WHERE department = 'QA';
```

This retrieves only the rows where:

```text
department = QA
```

# Row vs Column

| Row | Column |
|-----|--------|
| Represents a record | Represents an attribute |
| Horizontal | Vertical |
| Contains multiple values | Contains values of the same attribute |
| Example: one employee | Example: `salary` |
| Can be identified using a Primary Key | Has a name and data type |
