# 📘 Streamlit Interview Notes: `st.set_page_config()`

## 1. What is `st.set_page_config()`?

`st.set_page_config()` is used to configure the basic settings of a Streamlit web application page.

It allows you to control:

- Page title  
- Browser tab icon (favicon)  
- Page layout  
- Sidebar state  
- Menu behavior  

**Basic syntax:**
```python
import streamlit as st
st.set_page_config()
Example with options:

python
st.set_page_config(
    page_title="My App",
    page_icon="🤖",
    layout="wide"
)
2. Why do we use st.set_page_config()?
By default, Streamlit provides its own page title, layout, and settings.
Using st.set_page_config(), you can customize the page.

Example:

python
st.set_page_config(
    page_title="Data Science Dashboard",
    page_icon="📊",
    layout="wide"
)
Browser tab → 📊 Data Science Dashboard

3. Important Parameters
python
st.set_page_config(
    page_title=None,
    page_icon=None,
    layout="centered",
    initial_sidebar_state="auto",
    menu_items=None
)
4. page_title
Controls the title displayed in the browser tab.

python
st.set_page_config(page_title="Customer Churn Prediction")
Interview Point: Sets the browser tab title.

5. page_icon
Sets the favicon/icon displayed in the browser tab.

python
st.set_page_config(
    page_title="ML App",
    page_icon="🤖"
)
Interview Point: Specifies the favicon/icon.

6. layout
Controls the width of the main content area.

layout="centered" → Narrow, centered content

layout="wide" → Full-width content

Example:

python
st.set_page_config(layout="wide")
Use wide for: dashboards, charts, DataFrames, analytics apps.

7. initial_sidebar_state
Controls the initial state of the sidebar.

Options:

"expanded"

"collapsed"

"auto"

python
st.set_page_config(initial_sidebar_state="collapsed")
8. menu_items
Customize items in the Streamlit menu.

python
st.set_page_config(
    menu_items={
        "About": "This is my Data Science application."
    }
)
9. Complete Example
python
import streamlit as st

st.set_page_config(
    page_title="Customer Churn Prediction",
    page_icon="🤖",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.title("Customer Churn Prediction")
st.write("Welcome to the application!")
⚠️ Must be called before other Streamlit commands.

10. st.set_page_config() vs st.title()
st.set_page_config() → Configures the page itself (browser tab title, icon, layout).

st.title() → Displays a heading inside the app.

11. Interview Questions
Q1: What is st.set_page_config()?
A: Configures page settings like title, icon, layout, sidebar, menu.

Q2: Why do we use it?
A: To customize appearance and initial configuration.

Q3: What does page_title do?
A: Sets the browser tab title.

Q4: What does page_icon do?
A: Sets the favicon/icon.

Q5: Difference between centered and wide layout?
A: Centered = narrow container; Wide = full horizontal space.

Q6: When use layout="wide"?
A: Dashboards, charts, DataFrames, multi-column apps.

Q7: What does initial_sidebar_state do?
A: Controls sidebar visibility (expanded/collapsed/auto).

Q8: Where should st.set_page_config() be placed?
A: At the beginning of the script.

⭐ Quick Revision
page_title → Browser tab title

page_icon → Browser tab icon

layout → Centered / Wide

initial_sidebar_state → Auto / Expanded / Collapsed

menu_items → Customize Streamlit menu

✅ One-line Interview Answer
st.set_page_config() configures the Streamlit page’s metadata and layout, including the browser title, favicon, page width, sidebar state, and menu options.

================================

# 📦 Streamlit Interview Notes: `st.container()`

---

## What is a Container in Streamlit?

A **container** is a way to group multiple Streamlit elements together so you can control where those elements are displayed in your application.

You create a container using:

```python
st.container()
Think of a container as an empty box on your Streamlit page.

Example structure:

Code
Container
┌──────────────────────────────┐
│ Title                        │
│ Text                         │
│ Button                       │
│ DataFrame                    │
└──────────────────────────────┘
1. Basic Container Example
python
import streamlit as st

container = st.container()

container.title("My Application")
container.write("Welcome to my application")
container.button("Click Me")
Here, the three elements belong to the same container.

2. Why Use Containers?
Containers help organize your Streamlit UI.
For example, in a dashboard:

python
import streamlit as st

header = st.container()
header.title("Sales Dashboard")
header.write("Monthly Sales Analysis")

sales_section = st.container()
sales_section.metric("Total Sales", "₹50,000")
sales_section.metric("Total Orders", "250")
3. Using with st.container()
python
with st.container():
    st.title("Sales Dashboard")
    st.write("Welcome to the dashboard")
    st.button("View Sales")
This syntax is cleaner than assigning a variable.

4. Container with Multiple Elements
python
import streamlit as st

st.set_page_config(page_title="Sales Dashboard", layout="wide")

with st.container():
    st.title("Sales Dashboard")
    st.write("This is my sales dashboard")

    col1, col2 = st.columns(2)

    with col1:
        st.metric("Total Sales", "₹50,000")

    with col2:
        st.metric("Total Orders", "250")
Structure:

Container

Title

Description

Columns → Sales & Orders

5. Containers Can Be Updated
python
import streamlit as st
import time

container = st.container()
container.write("Processing...")

time.sleep(2)
container.write("Processing completed!")
Useful for ML predictions, progress updates, dashboards, chat apps, etc.

6. Container vs st.empty()
st.container() → Groups multiple elements

st.empty() → Single placeholder, replaceable later

python
placeholder = st.empty()
placeholder.write("Loading...")
placeholder.write("Completed!")
7. Container vs Columns
Container → Organizes elements vertically

Columns → Places elements side-by-side horizontally

python
col1, col2 = st.columns(2)

with col1:
    st.metric("Sales", "₹50K")

with col2:
    st.metric("Orders", "250")
You can combine containers and columns.

8. Container with Border
python
with st.container(border=True):
    st.subheader("Customer Information")
    st.write("Name: Shubham")
    st.write("Age: 31")
9. Container Height
python
with st.container(height=300):
    st.write("Content inside the container")
10. Container in GenAI Applications
python
import streamlit as st

with st.container():
    st.subheader("🤖 AI Assistant")
    question = st.text_input("Ask a question")

    if st.button("Ask"):
        st.write("AI Response:")
        st.write("This is the generated response.")
Interview Questions
Q1: What is a container in Streamlit?
A: A layout component used to group multiple UI elements.

Q2: How do you create a container?
A: container = st.container() or with st.container():

Q3: Why use st.container()?
A: To organize multiple elements into a section.

Q4: Difference between st.container() and st.empty()?
A: Container = multiple elements; Empty = single placeholder.

Q5: Difference between st.container() and st.columns()?
A: Container = vertical grouping; Columns = horizontal arrangement.

Q6: Can we put columns inside a container?
A: Yes, very common in dashboards.

Q7: Can a container have a border?
A: Yes → st.container(border=True)

Q8: Can we control container height?
A: Yes → st.container(height=300)

Q9: Where are containers useful?
A: Dashboards, ML apps, GenAI chatbots, forms, prediction results.

⭐ Quick Revision
st.container() → Group multiple elements

st.columns() → Arrange elements horizontally

st.empty() → Replaceable placeholder

✅ One-liner Interview Answer
st.container() is a Streamlit layout component used to group multiple UI elements into a specific section, making it easier to organize and manage the page layout.
