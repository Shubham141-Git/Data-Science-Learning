# MongoDB and MongoDB Atlas

## 1. What is MongoDB?

**MongoDB is a NoSQL database** that stores data in a **document-oriented format** instead of the traditional rows-and-columns format used by relational databases.

In relational databases such as **MySQL** or **PostgreSQL**, data is generally organized as:

```text
Database
   ↓
Tables
   ↓
Rows
   ↓
Columns

MongoDB organizes data as:

Database
   ↓
Collections
   ↓
Documents
   ↓
Fields
Example

In a relational database, an employee might look like:

id	name	department	salary
1	Shubham	IT	50000

In MongoDB, the same data can be represented as a document:

{
    "_id": 1,
    "name": "Shubham",
    "department": "IT",
    "salary": 50000
}

The document format is similar to JSON. Internally, MongoDB stores documents using BSON (Binary JSON).

3. What is a Document?

A document is a single record stored in MongoDB.

Example:

{
    "_id": 101,
    "name": "Rahul",
    "age": 25,
    "skills": ["Python", "SQL", "MongoDB"]
}

A document contains fields and their values.

Unlike a traditional relational table, MongoDB documents can have nested structures.

Example
{
    "name": "Rahul",
    "address": {
        "city": "Pune",
        "state": "Maharashtra"
    }
}

Here, address is a nested object containing city and state.

4. What is a Collection?

A collection is a group of MongoDB documents.

It is similar to a table in a relational database.

For example:

5. Why is MongoDB called NoSQL?

MongoDB is called a NoSQL database because it does not use the traditional relational table-and-row model as its primary data model.

Instead, MongoDB uses:

Collections
     ↓
Documents
     ↓
BSON

MongoDB is particularly useful when applications need flexible document structures and easy handling of nested data.

⭐ Interview Questions
Q1. What is MongoDB?
Answer

MongoDB is a NoSQL, document-oriented database that stores data as BSON documents inside collections.

Q2. Why is MongoDB called a NoSQL database?
Answer

MongoDB is called a NoSQL database because it does not use the traditional relational table-and-row model. It stores data as documents inside collections.

Q3. What is a document in MongoDB?
Answer

A document is a single record in MongoDB. It consists of fields and their values and is stored in BSON format.

Example
{
    "name": "Shubham",
    "age": 25,
    "city": "Pune"
}
Q4. What is a collection in MongoDB?
Answer

A collection is a group of MongoDB documents. It is roughly equivalent to a table in a relational database.

Q5. What is BSON?
Answer

BSON stands for Binary JSON. MongoDB uses BSON to store documents internally. BSON supports additional data types beyond standard JSON.

Q10. Does MongoDB have a primary key?
Answer

MongoDB automatically provides an _id field for each document if one is not supplied.

Example
{
    "_id": 101,
    "name": "Shubham"
}

The _id field uniquely identifies a document within its collection.

⭐ Important Interview Question
Q: Why would you choose MongoDB instead of a relational database?
Answer

MongoDB can be a good choice when the application works naturally with document-shaped data, requires flexible document structures, or needs to scale horizontally. Relational databases may be preferable when strong relational modeling, complex joins, or highly structured transactional data are central to the application.

# 🧩 Key Difference: Tables vs Collections

| Concept            | Relational Database (e.g., MySQL) | MongoDB (Non‑Relational) |
|--------------------|-----------------------------------|---------------------------|
| **Data Structure** | Tables (rows & columns)           | Collections (groups of documents) |
| **Record Unit**    | Row (tuple)                       | Document (JSON‑like object) |
| **Schema**         | Fixed — must define columns & types | Flexible — each document can have different fields |
| **Relationships**  | Foreign keys, joins               | Embedded documents or references |
| **Query Language** | SQL                               | MongoDB Query Language (MQL) |
| **Storage Format** | Structured tables                 | BSON (Binary JSON) documents |

---

## 🧠 Explanation

In a **relational database**, data is organized into **tables** — each table has a strict schema (columns, types, constraints).  

**Example (SQL):**
```sql
CREATE TABLE Students (
    id INT,
    name VARCHAR(50),
    age INT
);
In MongoDB, data is stored in collections, which are groups of documents. Each document is a JSON‑like structure:

json
{
  "id": 1,
  "name": "Shubham",
  "age": 22,
  "skills": ["Python", "Data Science"]
}
Collections don’t require a predefined schema — you can add new fields anytime.

Think of it like this:

Table → Collection

Row → Document

Column → Field

⚙️ Analogy
A table is like a spreadsheet — every row must have the same columns.

A collection is like a folder of JSON files — each file (document) can have unique keys.

💡 Why MongoDB uses Collections
Flexibility for evolving data structures

Easier to store nested or hierarchical data

Ideal for applications with dynamic or unstructured data (e.g., user profiles, logs, IoT data)

````python
# get all the smartphone category where price is greater than 0

query = {
    "category" : "smartphone",
    "price" : {"$gt" : 0}  # to check the price is greater than 0 we use $gt operator  and for less than we use $lt operator
    # "price" >  0
}
for i in collection.find(query):
    print(i)
````
