# Aggregate Functions in SQL

**Aggregate functions** are SQL functions used to **perform calculations on multiple rows and return a single summary value**.

> **Aggregate functions take multiple rows as input and produce a single summary result.**

### Example

Suppose we have the following salaries:

```text
50000
60000
70000
80000
```

Aggregate functions can calculate:

```text
SUM()   → 260000
AVG()   → 65000
MAX()   → 80000
MIN()   → 50000
COUNT() → 4
```

---

# Employee Table

We will use the following `employee` table:

| employee_id | name          | department | salary |
|------------:|---------------|------------|-------:|
| 1           | John Doe      | IT         | 50000  |
| 2           | Jane Smith    | HR         | 60000  |
| 3           | Alice Johnson | Finance    | 70000  |
| 4           | Bob Brown     | IT         | 80000  |

---

# Common Aggregate Functions

| Function | Purpose |
|----------|---------|
| `COUNT()` | Counts rows or non-NULL values |
| `SUM()` | Calculates the total |
| `AVG()` | Calculates the average |
| `MAX()` | Finds the maximum value |
| `MIN()` | Finds the minimum value |

---

# 1. COUNT()

`COUNT()` is used to **count rows or values**.

## Count All Employees

```sql
SELECT COUNT(*)
FROM employee;
```

### Result

```text
4
```

There are **4 employees**.

## Count Employees with Salary Values

```sql
SELECT COUNT(salary)
FROM employee;
```

This counts only the **non-NULL values** in the `salary` column.

> **Important:** `COUNT(*)` counts all rows, while `COUNT(column)` counts only non-NULL values in that column.

---

# 2. SUM()

`SUM()` calculates the **total of a numeric column**.

## Find Total Salary

```sql
SELECT SUM(salary)
FROM employee;
```

### Calculation

```text
50000 + 60000 + 70000 + 80000
                    ↓
                  260000
```

### Result

```text
260000
```

---

# 3. AVG()

`AVG()` calculates the **average value**.

## Find Average Salary

```sql
SELECT AVG(salary)
FROM employee;
```

### Calculation

```text
(50000 + 60000 + 70000 + 80000) / 4
                    ↓
                  65000
```

### Result

```text
65000
```

---

# 4. MAX()

`MAX()` returns the **largest value**.

## Find the Highest Salary

```sql
SELECT MAX(salary)
FROM employee;
```

### Result

```text
80000
```

---

# 5. MIN()

`MIN()` returns the **smallest value**.

## Find the Lowest Salary

```sql
SELECT MIN(salary)
FROM employee;
```

### Result

```text
50000
```

---

# Using Multiple Aggregate Functions

We can use multiple aggregate functions in the same query.

```sql
SELECT
    COUNT(*) AS total_employees,
    SUM(salary) AS total_salary,
    AVG(salary) AS average_salary,
    MAX(salary) AS highest_salary,
    MIN(salary) AS lowest_salary
FROM employee;
```

### Result

| total_employees | total_salary | average_salary | highest_salary | lowest_salary |
|----------------:|-------------:|---------------:|---------------:|--------------:|
| 4               | 260000       | 65000          | 80000          | 50000         |

This is useful for **data analysis and reporting**.

---

# Aggregate Functions with WHERE

`WHERE` can be used to **filter rows before performing the aggregate calculation**.

## Find Total Salary of IT Employees

```sql
SELECT SUM(salary) AS total_IT_salary
FROM employee
WHERE department = 'IT';
```

IT employees:

```text
John Doe  → 50000
Bob Brown → 80000
```

### Calculation

```text
50000 + 80000
     ↓
   130000
```

### Result

```text
130000
```

## Find Average IT Salary

```sql
SELECT AVG(salary) AS average_IT_salary
FROM employee
WHERE department = 'IT';
```

### Result

```text
65000
```

---

# Aggregate Functions with GROUP BY

This is **very important for SQL interviews**.

`GROUP BY` allows us to calculate aggregate values **for each group**.

## Find Total Salary for Each Department

```sql
SELECT
    department,
    SUM(salary) AS total_salary
FROM employee
GROUP BY department;
```

### Result

| department | total_salary |
|------------|-------------:|
| Finance    | 70000        |
| HR         | 60000        |
| IT         | 130000       |

Instead of calculating one total for the entire table, we calculate **one total for each department**.

### Visualization

```text
                 employee table
                       │
                       ▼
                  GROUP BY
                       │
          ┌────────────┼────────────┐
          ▼            ▼            ▼
       Finance         HR           IT
          │            │            │
          ▼            ▼            ▼
       SUM(salary)  SUM(salary)  SUM(salary)
          │            │            │
          ▼            ▼            ▼
        70000        60000        130000
```

---

# WHERE → GROUP BY → Aggregate

A common SQL processing flow is:

```text
┌──────────────────┐
│  Employee Table  │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│      WHERE       │
│   Filter Rows    │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│     GROUP BY     │
│   Create Groups  │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│ Aggregate        │
│ Function         │
│ SUM / AVG / etc. │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│  Summary Result  │
└──────────────────┘
```

### Example

```sql
SELECT
    department,
    AVG(salary) AS average_salary
FROM employee
WHERE salary > 50000
GROUP BY department;
```

---

# Aggregate Function vs Normal Function

## Normal Function

A normal function generally works with **individual values**.

```text
Individual Value
       ↓
     Function
       ↓
    Result
```

## Aggregate Function

An aggregate function works with **multiple rows** and produces a summary.

```text
Multiple Rows
      ↓
Aggregate Function
      ↓
Single Summary Value
```

---

# WHERE vs HAVING

This is a **very important SQL interview question**.

| Clause | Purpose |
|--------|---------|
| `WHERE` | Filters individual rows |
| `GROUP BY` | Creates groups |
| Aggregate Function | Calculates values for groups |
| `HAVING` | Filters groups after aggregation |

### Example

```sql
SELECT
    department,
    AVG(salary) AS avg_salary
FROM employee
GROUP BY department
HAVING AVG(salary) > 60000;
```

Here:

```text
Employee Table
      ↓
GROUP BY department
      ↓
Calculate AVG(salary)
      ↓
HAVING AVG(salary) > 60000
      ↓
Return matching departments
```

```sql
SELECT AVG(salary)
FROM employee
WHERE department = 'IT';
```

```text
All Employees
     ↓
WHERE department = 'IT'
     ↓
IT Employees
     ↓
AVG(salary)
     ↓
65000
```

## Q5. Can Aggregate Functions be used with `GROUP BY`?

### Answer

Yes. This is one of the most common uses of aggregate functions.

---

## Q6. What is the difference between `WHERE` and `HAVING`?

### Answer

> `WHERE` filters individual rows **before grouping**, while `HAVING` filters groups **after aggregation**.

### Example

```sql
SELECT
    department,
    AVG(salary) AS avg_salary
FROM employee
GROUP BY department
HAVING AVG(salary) > 60000;
```
| Feature                          | `WHERE`                      | `HAVING`                      |
| -------------------------------- | ---------------------------- | ----------------------------- |
| **Purpose**                      | Filters individual rows      | Filters groups                |
| **Used before/after `GROUP BY`** | Used **before** `GROUP BY`   | Used **after** `GROUP BY`     |
| **Works on**                     | Individual rows              | Grouped/aggregated results    |
| **Can use aggregate functions?** | Generally **No**             | **Yes**                       |
| **Commonly used with**           | `SELECT`, `UPDATE`, `DELETE` | `GROUP BY`                    |
| **Example condition**            | `salary > 50000`             | `COUNT(*) > 2`                |
| **Question it answers**          | "Which rows should I keep?"  | "Which groups should I keep?" |

# Nested Query / Subquery in SQL

A **subquery** is a SQL query written inside another SQL query.

In simple words:

> A subquery is a query inside a query.

The inner query generally executes first, and its result is used by the outer query.

---

## Basic Structure

```sql
SELECT column_name
FROM table_name
WHERE column_name = (
    SELECT column_name
    FROM table_name
    WHERE condition
);
```

## Logical Flow

```text
Inner Query (Subquery)
        ↓
Inner Query Result
        ↓
Outer Query uses the result
        ↓
Final Result
```

---

## Example Employee Table

We will use the following table:

| employee_id | name          | department | salary |
|---:|---------------|------------|-------:|
| 1 | John Doe      | IT         | 50000 |
| 2 | Jane Smith    | HR         | 60000 |
| 3 | Alice Johnson | Finance    | 70000 |
| 4 | Bob Brown     | IT         | 80000 |

---

# Example 1: Find the Employee with the Highest Salary

## Question

Find the employee who has the highest salary.

First, find the highest salary:

```sql
SELECT MAX(salary)
FROM employee;
```

**Result:**

```text
80000
```

Now find the employee whose salary is `80000`:

```sql
SELECT *
FROM employee
WHERE salary = 80000;
```

Instead of running two separate queries, use a subquery:

```sql
SELECT *
FROM employee
WHERE salary = (
    SELECT MAX(salary)
    FROM employee
);
```

## How Does It Work?

The inner query runs first:

```sql
SELECT MAX(salary)
FROM employee;
```

It returns:

```text
80000
```

The outer query then becomes conceptually:

```sql
SELECT *
FROM employee
WHERE salary = 80000;
```

## Flow

```text
Employee Table
      ↓
Inner Query: Find MAX(salary)
      ↓
    80000
      ↓
Outer Query: Find employee with salary = 80000
      ↓
  Bob Brown
```

---

Now find employees whose salary is greater than `65000`:

```sql
SELECT *
FROM employee
WHERE salary > 65000;
```

Using a subquery:

# Example 4: Using a Subquery with `IN`

A subquery does not always return a single value. It can also return multiple values.

# `=` vs. `IN` with Subqueries

This distinction is important.

## Use `=` When the Subquery Returns One Value

```sql
SELECT *
FROM employee
WHERE salary = (
    SELECT MAX(salary)
    FROM employee
);
```

The subquery returns:

```text
80000
```

It returns one value, so `=` is appropriate.

## Use `IN` When the Subquery Can Return Multiple Values

```sql
SELECT *
FROM employee
WHERE department IN (
    SELECT department
    FROM department
    WHERE location = 'Mumbai'
);
```

The subquery could return multiple values:

```text
IT
HR
Finance
```

For multiple possible values, use `IN`.

| Operator | Use case |
|----------|----------|
| `=` | The subquery returns one value |
| `IN` | The subquery can return multiple values |

---

# Example 5: Subquery in the `FROM` Clause

A subquery can also be placed inside the `FROM` clause.

## Question

Find the average of the department-level total salaries.

```sql
SELECT AVG(total_salary)
FROM (
    SELECT department, SUM(salary) AS total_salary
    FROM employee
    GROUP BY department
) AS department_salary;
```

## How It Works

```text
Inner Query
     ↓
Calculate total salary for each department
     ↓
Department-level result
     ↓
Outer Query
     ↓
Calculate the average of the department totals
```

A subquery inside the `FROM` clause is often called a **derived table**.

> A subquery in `FROM` should normally have an alias, such as `department_salary`.

---

# Example 6: Subquery with `HAVING`

## Question

Find departments whose average salary is greater than the overall average salary.

```sql
SELECT
    department,
    AVG(salary) AS average_salary
FROM employee
GROUP BY department
HAVING AVG(salary) > (
    SELECT AVG(salary)
    FROM employee
);
```

This query combines:

- `GROUP BY`
- `AVG()`
- `HAVING`
- A subquery

This type of query is useful in SQL interviews.

# Types of Subqueries

| Type | Description |
|------|-------------|
| **Single-row subquery** | Returns one row |
| **Multiple-row subquery** | Returns multiple rows |
| **Scalar subquery** | Returns exactly one value |
| **Correlated subquery** | Depends on values from the outer query |
| **Subquery in `WHERE`** | Filters rows |
| **Subquery in `FROM`** | Acts like a temporary or derived table |
| **Subquery in `SELECT`** | Calculates a value for each result row |

# Interview Questions

## Q1. What Is a Subquery?

**Answer:**

A subquery is a SQL query nested inside another SQL query. The result of the inner query is used by the outer query.

---

## Q2. Which Query Executes First?

**Answer:**

For a **non-correlated subquery**, the inner query is generally evaluated before the outer query.

Example:

```sql
SELECT *
FROM employee
WHERE salary > (
    SELECT AVG(salary)
    FROM employee
);
```

The inner query is evaluated first:

```sql
SELECT AVG(salary)
FROM employee;
```

> In a correlated subquery, the inner query depends on the current row of the outer query and may be evaluated repeatedly.

---

## Q3. What Is the Difference Between `IN` and `=`?

**Answer:**

- Use `=` when the subquery returns a single value.
- Use `IN` when the subquery can return multiple values.

### Single Value

```sql
WHERE salary = (
    SELECT MAX(salary)
    FROM employee
);
```

### Multiple Values

```sql
WHERE department IN (
    SELECT department
    FROM department
);
```

---

## Q4. Can a Subquery Be Used with `WHERE`?

**Answer:**

Yes. A subquery can provide a value or set of values used to filter rows.

```sql
SELECT *
FROM employee
WHERE salary > (
    SELECT AVG(salary)
    FROM employee
);
```

---

## Q5. Can a Subquery Be Used with `FROM`?

**Answer:**

Yes. A subquery inside `FROM` acts like a temporary result set or derived table.

```sql
SELECT *
FROM (
    SELECT
        department,
        SUM(salary) AS total_salary
    FROM employee
    GROUP BY department
) AS result;
```

# Most Important Interview Question: Find the Second-Highest Salary

```sql
SELECT MAX(salary) AS second_highest_salary
FROM employee
WHERE salary < (
    SELECT MAX(salary)
    FROM employee
);
```

## How It Works

The inner query finds the highest salary:

```sql
SELECT MAX(salary)
FROM employee;
```

**Result:**

```text
80000
```

The outer query considers only salaries lower than `80000`:

```text
50000
60000
70000
```

It then applies `MAX()` to the remaining salaries:

```text
70000
```

## Answer

```text
70000
```

> This query finds the second-highest **distinct** salary. If the highest salary appears more than once, those duplicate highest salaries are excluded before calculating the next highest value.
