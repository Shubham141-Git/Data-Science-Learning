# Subplots: Multiple charts in one figure

# Let’s compare monthly revenue vs monthly quantity

```python
df_quan
```

```python
monthly_quantity = df_quan.groupby('Month')['Quantity'].sum().sort_index()

monthly_revenue = df_quan.groupby('Month')['Revenue'].sum()
```

```python
fig, axs = plt.subplots(nrows=2, ncols=1, figsize=(10, 8), sharex=True) # just understand this rest is asame as above

# Revenue Plot
axs[0].plot(monthly_revenue.index, monthly_revenue.values, marker='o', color='blue', label='Revenue-label')
axs[0].set_title("Monthly Revenue")
axs[0].set_xlabel("Month")
axs[0].set_ylabel("Revenue")
axs[0].legend()  # it wil show the label
axs[0].grid(True)

# Quantity Plot
axs[1].plot(monthly_quantity.index, monthly_quantity.values, marker='s', color='green', label='Quantity-label')
axs[1].set_title("Monthly Quantity")
axs[1].set_xlabel("Month")
axs[1].set_ylabel("Quantity")
axs[0].legend() 
axs[1].grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# How to Show the labels in the chart
fig, axs = plt.subplots(nrows=2, ncols=1, figsize=(10, 8), sharex=True)

# Revenue Plot
line1, = axs[0].plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker='o',
    color='blue',
    label='Revenue'
)

# Quantity Plot
line2, = axs[1].plot(
    monthly_quantity.index,
    monthly_quantity.values,
    marker='s',
    color='green',
    label='Quantity'
)

# Show both labels in the upper graph
axs[0].legend(handles=[line1, line2])

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Seaborn: High-Level Statistical Data Visualization

Seaborn is a powerful Python visualization library built on top of Matplotlib and integrated with pandas. It simplifies many common tasks like:

- Statistical plots

- Categorical data plots

- Distribution plots

- Color themes

- Multi-variable analysis

```python
# New Analysis: Distribution of Quantity per Transaction

df_quantity = df_quan[(df_quan['Quantity'] > 0) & (df['Quantity'] < 100)] # here we r filtering out the records of quantitty range between 1 to 100

#fter filtering we will ahave 100 points on x axis ,but we can divide 100 points into 10 part like 0-10, 11-20 this are called bins

# 1 to 99

df_quantity
```

#### Show me how transaction quantities (between 1 and 99) are distributed. Bin them into 10 groups, but also draw a smooth probability curve showing where quantity values tend to cluster.


```python
import seaborn as sns
plt.figure(figsize=(12, 6))
sns.histplot(df_quantity['Quantity'], bins=10, kde=False, color='skyblue')
plt.title('Distribution of Revenue')
plt.xlabel('Quantity bins')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# Kernel Density Estimate
```

```python
#when kde is true kde  tell us the  / probablity of  getting rhe invoice between each bin

'''  
KDE stands for Kernel Density Estimation.

It is used to estimate the probability density (distribution) of continuous data. Instead of showing bars like a histogram, 
KDE draws a smooth curve that represents how the data is distributed.
'''


plt.figure(figsize=(12, 6))
sns.histplot(df_quantity['Quantity'], bins=10, kde=True, color='skyblue')
plt.title('Distribution of Revenue')
plt.xlabel('Quantity bins')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# Kernel Density Estimate
```

```python
# sns.kdeplot() — KDE (smooth version of histogram)
plt.figure(figsize=(12, 8))
sns.kdeplot(df_quantity['Quantity'], fill=True, color='green')
plt.title('Revenue KDE Curve')
plt.xlabel('Revenue')
plt.tight_layout()
plt.show()
```
