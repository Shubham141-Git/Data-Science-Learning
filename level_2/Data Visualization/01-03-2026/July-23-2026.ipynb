# What is Matplotlib?

- Core plotting engine behind libraries like Seaborn and Pandas plots.

```python
import matplotlib.pyplot as plt
```

```python
# Mow to get the monthy revenue

# here we can say wrt to data frame that each month willl have the combine revenue of different  invoice
# So we need to group the all the invoice of all the months and Sum thier months
```

```python
# - group by month
# - Sum their revenue



df_quan['Month'] = df_quan['InvoiceDate'].dt.to_period('M').astype(str)
df_quan.head()
```

```python
# Get monthly reveneue
# Step-1  --> Group by Month
# Step-2 --> Sum thier revenue

df_quan['Revenue'] = df_quan['Quantity'] * df_quan['Price']
```

```python
dfGroupMonth=df_quan.groupby("Month")
dfGroupMonth.head()
```

```python
Monthly_Revenue=df_quan.groupby("Month")["Revenue"].sum().sort_index()
Monthly_Revenue
```

```python
# Create A CAnvas
plt.Figure(figsize=(12,6))
plt.style.use('ggplot') # Used to make fancy
#Plot
plt.plot(Monthly_Revenue.index, Monthly_Revenue.values, marker='o') # here x axis will be the index of the datarame and its values wll the y axis values
# | Marker Symbol | Description     |
# | ------------- | --------------- |
# | `'o'`         | Circle          |
# | `'.'`         | Point           |
# | `','`         | Pixel           |
# | `'x'`         | X               |
# | `'+'`         | Plus            |
# | `'s'`         | Square          |
# | `'D'`         | Diamond         |
# | `'^'`         | Upward triangle |

#Title --> Monthly revenue over time
plt.title('Monthly Revneue Over Time')

# X-axis label
plt.xlabel("Month")


# Y-axis label
plt.ylabel("Revenue")

#rotate X labels

plt.xticks(rotation=90)

# Fit layout, prevents label clipping
plt.tight_layout(pad=0.7)

# Show grid
plt.grid(True) 

# show the plot window or render the figure
plt.show()
```
